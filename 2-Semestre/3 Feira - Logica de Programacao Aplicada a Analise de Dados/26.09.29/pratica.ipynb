### Registro de participantes

```python
def registrar_treinamento(tema, *participantes):

    print(f"O tema do treinamento é: {tema}")
    print(f"Há {len(participantes)} participante(s) neste treinamento")

    for numero, participante in enumerate(participantes, start=1):
        print(f"{numero}° Participante: {participante}")
```

```python
registrar_treinamento("SQL", "João", "Ana", "Leticia", "Leo", "Vitor")
```

```python
registrar_treinamento("SQL", "João")
```

## Problema 1 - Problemas de vendas por vendedor

```python
def analisa_vendas(vendedor, *vendas):
    soma_vendas = 0
    for i in vendas:
        soma_vendas += i
    
    media_vendas = soma_vendas/len(vendas)
    
    maior_venda = max(vendas)
    
    if soma_vendas >= 5000:
        meta = "Meta atingida"
    else:
        meta = "Meta não atingida"
    
    print(f"Vendedor: {vendedor}")
    print(f"Valor total vendido: R$ {soma_vendas}")
    print(f"Média das vendas: {media_vendas}")
    print(f"Maior valor de venda: {maior_venda}")
    print(meta)
    print(50*"-")
```

```python
analisa_vendas("José", 500, 600, 500, 2000, 1500, 100, 900, 400)
analisa_vendas("Pedro", 600, 600, 500, 1960, 1500)
analisa_vendas("Lucas", 500, 600, 500, 1400)
```

## Problema 2 - Registro de atendimento

```python
def registrar_atendimento(nome, *procedimentos, **informacoes):
    
    print(f"Nome do cliente: {nome}") # Nome do cliente
    print(f"Quantidade de procedimentos: {len(procedimentos)}") # Quantidade de procedimentos
    
    print(50*"-")
    
    for numero, procedimento in enumerate(procedimentos, start=1): # Mostra os procedimentos
        print(f"{numero}° Procedimento: {procedimento}")
        
    print(50*"-")
    
    if informacoes != {}: # Verifica se há informações adicionais
        print("Informações adicionais:") # Se tiver
        for chave, valor in informacoes.items():
            if chave != "retorno": 
                print(f"{chave}: {valor}")
    else: # Se não tiver
        print("Sem informações adicionais")

    print(50*"-")

    if "retorno" in informacoes: # Se tiver mostra ou diz que não tem data de retorno
        print(f"Data de retorno: {informacoes['retorno']}")
    else:
        print("Retorno não informado")
        
```

```python
registrar_atendimento("Lapa", "Botox", "Amputação de dedo", "Bariatrica", retorno="12/10/2026", responsavel="Doutor Kevin", convenio=True)
```

```python
registrar_atendimento("Rodrigo", "Mastodonte")
```

```python
registrar_atendimento("Saci", "Amputação de perna", retorno="24/12/2026", responsavel="Doutor Cauan", convenio=False)
```

## Prática - Cadastro funcionário

```python
def cadastrar_funcionario(nome, idade, salario, setor):
    
    if not nome:
        print("O nome é obrigatório!")
        return

    if idade < 18 or idade > 70:
        print("Idade inválida")
        return

    if salario < 0:
        print("Salário inválido")
        return
    
    if setor not in ["TI", "RH", "Financeiro", "Comercial"]:
        print("Setor inválido")
        return
    
    print("Funcionário cadastrado com sucesso")
```

```python
cadastrar_funcionario("", 52, 1800, "TI")
cadastrar_funcionario("João", 17, 1500, "RH")
cadastrar_funcionario("Lapa", 21, -5000, "Marketing")
cadastrar_funcionario("Lapa", 21, 5000, "Marketing")
cadastrar_funcionario("Lapa", 21, 5000, "TI ")
```
