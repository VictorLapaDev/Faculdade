### Como são muitas infromações nulas e muitas vezes por ser uma jogada que geralmente não acontece, se excluirmos esses dados vamos exluir grande parte dos nossos dados, por isso vamos ransformar em tipos categóricos ou strings

```python
colunas_jogadores_nulos = [
    'RecFumbPlayer', 'Interceptor', 'Returner', 'PenalizedPlayer', 
    'Tackler2', 'Rusher', 'Rusher_ID', 'Receiver_ID', 'Receiver', 
    'Passer', 'Passer_ID', 'BlockingPlayer'
]
df[colunas_jogadores_nulos] = df[colunas_jogadores_nulos].fillna('Nenhum').astype('string')

colunas_categorias_especificas = [
    'DefTwoPoint', 'TwoPointConv', 'ChalReplayResult', 'RecFumbTeam', 
    'FieldGoalResult', 'ExPointResult', 'PenaltyType', 'Timeout_Team', 
    'ReturnResult', 'PuntResult', 'PenalizedTeam', 'RunGap', 
    'RunLocation', 'PassLocation', 'PassLength', 'PassOutcome'
]
for coluna in colunas_categorias_especificas:
    df[coluna] = df[coluna].astype('category')
    
    if 'Nenhum' not in df[coluna].cat.categories:
        df[coluna] = df[coluna].cat.add_categories('Nenhum')
        
    df[coluna] = df[coluna].fillna('Nenhum')
```

```python
# Arrumar colunas numéricas

# Garantir o tipo numérico otimizado preservando os nulos
df['FieldGoalDistance'] = df['FieldGoalDistance'].astype('Int64')

cols_epa_wpa = ['airEPA', 'yacEPA', 'airWPA', 'yacWPA']
df[cols_epa_wpa] = df[cols_epa_wpa].astype('float64')
```

📝 Justificativa: Tratamento de Métricas Numéricas Condicionais
As variáveis FieldGoalDistance, airEPA, yacEPA, airWPA e yacWPA são métricas numéricas condicionais:

FieldGoalDistance: Só é aplicável quando há uma tentativa explicita de chute de campo (Field Goal).

airEPA/yacEPA/airWPA/yacWPA: Correspondem à decomposição do impacto de pontos/vitória específicos para passes completos/incompletos.

Decisão de tratamento: Mantiveram-se os valores nulos (NaN / <NA>) para evitar distorções estatísticas que ocorreriam com imputação por zero, média ou mediana. Os tipos de dados foram otimizados para Int16 e float32, garantindo máxima compressão no formato Parquet sem alterar a integridade matemática das informações.
