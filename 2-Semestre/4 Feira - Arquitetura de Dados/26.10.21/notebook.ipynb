# Tratamento dos Tipos de Dados Coluna por Coluna

```python
df.dtypes.value_counts()
```

```python
missing_tokens = ["NA", "N/A", "null", "None", "nan", "UNKNOWN"]

df.isin(missing_tokens).sum()
```

```python
data_dict_object = pd.DataFrame({  # cria um DataFrame resumo (dicionário de dados) apenas para colunas do tipo object

    "variavel": df.columns,  # adiciona uma coluna com os nomes das variáveis do tipo object

    "Amostras (em ordem alfabética)": [", ".join(sorted(df[col].dropna().astype(str).unique()))  # lista os valores únicos da coluna, ordenados alfabeticamente e separados por vírgula
                                       if df[col].notna().any() else None  # se a coluna tiver apenas NaN, retorna None
                                       for col in df.columns],  # repete esse processo para cada coluna object

    "Amostras (em ordem de frequência)": [", ".join(df[col].dropna().astype(str).value_counts().index.tolist())  # lista os valores únicos ordenados pela frequência (mais comum primeiro)
                                          if df[col].notna().any() else None  # se só tiver NaN, retorna None
                                          for col in df.columns],  # repete para cada coluna object
    
    "qtd_valores_unicos": [df[col].nunique()  # calcula a quantidade de valores únicos em cada coluna
                           for col in df.columns]

})

# Colocar em ordem alfabética
data_dict_object = data_dict_object.sort_values(
    by="Amostras (em ordem alfabética)",  # ordena o DataFrame com base nos valores das amostras em ordem alfabética
    ascending=True,  # ordem crescente
    na_position="last"  # coloca os valores None no final
).reset_index(drop=True)  # reseta o índice após a ordenação

data_dict_object  # exibe o DataFrame final com o resumo das colunas do tipo object
```
