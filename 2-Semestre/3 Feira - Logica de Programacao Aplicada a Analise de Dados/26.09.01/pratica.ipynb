```python
def texto_colorido(texto, cor = "branco", estilo = "normal"):
    cores = {
        "vermelho": "\033[31m",
        "verde": "\033[32m",
        "amarelo": "\033[33m",
        "azul": "\033[34m",
        "magenta": "\033[35m",
        "ciano": "\033[36m",
        "branco": "\033[37m"
    }
    
    estilos = {
        "normal": "",
        "negrito": "\033[1m",
        "italico": "\033[3m",
        "sublinhado": "\033[4m"
    }

    if cor in cores and estilo in estilos:
        return f"{estilos[estilo]}{cores[cor]}{texto}\033[0m"
    else:
        return texto  # Retorna o texto sem formatação se a cor ou estilo não forem válidos
```

```python
texto = texto_colorido("Olá, mundo!", "vermelho", "sublinhado")
print(texto)
```

# Medidas Centrais

São: Média(Todos valores do conjunto), Moda(Maior frequencia), Mediana(Valor central dos dados ordenados)

Modal, multimodal, bimodal, amodal

```python
def calcular_mediana(lista):

    lista = sorted(lista)
    tamanho = len(lista)
    meio = tamanho // 2

    if tamanho % 2 == 0: # Par

        elemento_1 = lista[meio]
        elemento_2 = lista[meio-1]

        mediana = (elemento_1 + elemento_2) / 2

        return mediana
        
    else: # Impar     
        return lista[meio] 
```

```python
medianaA = calcular_mediana([23, 14, 31, 17, 20]) # 14, 17, 20, 23, 31   20
print(medianaA)

print("-------------------")

medianaB = calcular_mediana([75, 1, 4, 3, 19, 50, 64, 31])
print(medianaB)
```

```python
def calcular_moda(lista):
    moda = None
    frequencia = 0
    for i in lista:
        if lista.count(i) > frequencia:
            frequencia = lista.count(i)
            moda = i
    return moda
```

```python
modaA = calcular_moda([23, 14, 31, 17, 23, 20, 23, 31])
print(modaA)
```

```python
def calcular_moda(lista):
    modas = []
    maior_frequencia = 0

    for valor in lista:
        frequencia = lista.count(valor)

        if frequencia > maior_frequencia:
            maior_frequencia = frequencia
            modas = [valor]

        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)


    if maior_frequencia == 1:
        return "Amodal"
    elif len(modas) == 1:
        return "Unimodal", modas
    elif len(modas) == 2:
        return "Bimodal", modas
    else:
        return "Multimodal", modas
```

```python
dados_1 = [10, 20, 20, 30, 40]
dados_2 = [10, 10, 20, 20, 30]
dados_3 = [10, 10, 20, 20, 30, 30, 40]
dados_4 = [10, 20, 30, 40]

print(calcular_moda(dados_1))
print(calcular_moda(dados_2))
print(calcular_moda(dados_3))
print(calcular_moda(dados_4))
```
