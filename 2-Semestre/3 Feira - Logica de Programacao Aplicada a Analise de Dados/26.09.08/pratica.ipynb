## Medidas de Dispersão

### Variância

O quanto os dados estão espalhados ao quadrado.

**IMPORTANTE:** Não comparar aos dados originais 

![Variância](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRY7lJblIt7zhjvKF-y_s8vpoKAzlZqgi-NgPswhvZjJYGL0f-KXfg5oJQ&s=10)

### Desvio Padrão 

Medida de dispersão que indica o quanto os valores de um conjunto de dados estão dispersos em relação à média.

**IMPORTANTE:** A sua vantagem é que ele retorna a variância para a mesma unidade dos dados originais 

### Amplitude

Diferença entre o valor máximo e mínimo do conjunto de dados
