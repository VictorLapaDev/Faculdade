```python
def eh_primo(num):
    # Valida se o número é 0
    if num == 0:
        return False
        
    # Valida de 2 até um número antes do valor incerido  
    #Ele divide o numero por todos os números até ele
    #Como todo primo só é divisivel por 1 e ele mesmo, se algum valor der 0, ele pula da função retornando false
    #Se nenhum der como resto 0, ele retorna como True
    for x in range(2, num):
        if num % x == 0:
            return False
    return True
```

```python
def gerar_escada(num):
    # Percorre cada linha da escada (de 0 até num-1)
    for x in range(num):
        texto = ""  # Inicializa a string que armazenará os caracteres da linha atual
        
        # Percorre as colunas da esquerda para a direita (de 0 até num-1)
        for y in range(num):
            
            # Se o número da coluna (y) for menor ou igual ao número da linha (x),
            # adiciona "*" à string.
            # Caso contrário, adiciona um espaço em branco.
            # Isso faz com que os asteriscos aumentem a cada linha,
            # formando uma escada crescente.
            if y <= x:
                texto += "*"
            else:
                texto += " "
        
        # Exibe a linha completa após montar todos os caracteres
        print(texto)
```

```python
gerar_escada(9)
```

### Exercício 23

```python
def interseccao_listas(lista1, lista2):
    nova_lista = []
    
    posicao = 0
    while posicao < len(lista1):
        
        for i in lista2:
            #Verifica se há nas duas listas
            if lista1[posicao] == i:
                # Verifica se ja existe na nova lista para nao adicionar novamente
                existe = False
                for x in nova_lista:
                    if i == x:
                        existe = True
                        break
                if not existe:
                    nova_lista.append(i)
                    break
                
        posicao += 1
    return nova_lista
```

```python
vetor = [9, 34, 65, 34, 6, 87,9, 6, 90, 9]
lista_ordenada = [2, 4, 6, 7, 9, 12, 9, 67]

ambas_listas = interseccao_listas(vetor, lista_ordenada)
print(ambas_listas)
```

### Exercício 27

```python
def selection_sort(lista):
    
    for i in range(len(lista)): # Percorre a lista
        menor_valor=i # Define o primeiro valor lido como o menor(pega o indice)
        
        for j in range(i +1, len(lista)): # Percorre do valor verificado anteriormente para a frente na lista
            if lista[j] < lista[menor_valor]: # Se o valor posterior for menor que o já lido, ele define ele como o menor, e executa até o final da lista, procurando o menor valor
                menor_valor = j # Salva o indice do menor numero
                
        var_aux = lista[i] # Variavel temporaria para nao perder o valor do elemento que estamos percorrendo com I
        lista[i] = lista[menor_valor] # Define o menor número na frente da lista
        lista[menor_valor] = var_aux # Define o valor que foi substituido no lugar do que foi trocado
        
    return lista
        
```

```python
vetor = [9, 34,-65, 34, 6, 8,0, 7,9, 6, 90, 9, 9]

lista_sort = selection_sort(vetor)
print(lista_sort)
```

### Exercício 28

```python
def remove_first_last(lista):
    lista.pop(len(lista)-1)
    lista.pop(0)
    return lista
```

```python
vetor = [9, 34, 65, 34, 6, 87,9, 6, 90, 9, 50]

new_lista = remove_first_last(vetor)
print(new_lista)
```

### Exercício 30

```python
def somar_listas(lista1, lista2):
    if len(lista1) != len(lista2):
        return "Listas de tamanhos diferentes"
    else:
        nova_lista = []
        for i in range(len(lista1)):
            nova_lista.append(lista1[i] + lista2[i])
        return nova_lista
```

```python
vetor = [9, 34, 65, 34, 6, 87,9, 6, 90, 9]
lista_ordenada = [2, 4, 6, 7, 9, 12, 9, 67, 28, 20]

soma = somar_listas(vetor, lista_ordenada)
print(soma)
```
