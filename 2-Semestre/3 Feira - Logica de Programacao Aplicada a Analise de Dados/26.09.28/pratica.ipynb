# Atividade Prática 1

```python
chamados = [
{
"codigo": "CH001",
"responsavel": "Maria",
"abertura": "15/09/2026",
"fechamento": "17/09/2026",
"prioridade": "Alta"
},
{
"codigo": "CH002",
"responsavel": "Carlos",
"abertura": "16/09/2026",
"fechamento": "22/09/2026",
"prioridade": "Média"
},
{
"codigo": "CH003",
"responsavel": "Marina",
"abertura": "18/09/2026",
"fechamento": "19/09/2026",
"prioridade": "Alta"
},
{
"codigo": "CH004",
"responsavel": "Rafael",
"abertura": "20/09/2026",
"fechamento": "25/09/2026",
"prioridade": "Baixa"
},
{
"codigo": "CH005",
"responsavel": "Maria",
"abertura": "21/09/2026",
"fechamento": "23/09/2026",
"prioridade": "Média"
}
]
```

## Parte 1

```python
from datetime import datetime
```

```python
for numero, chamado in enumerate(chamados, start=1): # Transforma as datas de string para datetime
    try:
        chamado["abertura"] = datetime.strptime(chamado["abertura"], "%d/%m/%Y")
        chamado["fechamento"] = datetime.strptime(chamado["fechamento"], "%d/%m/%Y")
        
        chamado["tempo_chamado_d"] = (chamado["fechamento"] - chamado["abertura"]).days # Gera uma chave no dicionario
        
        if chamado["tempo_chamado_d"] <= 3:
            chamado["prazo"] = "Dentro do prazo"
        else:
            chamado["prazo"] = "Fora do prazo"
            
    except:
        print(f"{numero} - {chamado["codigo"]} | {chamado["responsavel"]} | {chamado["prioridade"]} | {chamado["tempo_chamado_d"]} Dias | {chamado["prazo"]}")        
        print(50*"-")
```

## Parte 2

```python
in_prazo = 0
out_prazo = 0
chamado_maior_temp = max(chamados, key=lambda x: x["tempo_chamado_d"])
quant_alto = 0
quant_medio = 0
quant_baixo = 0

for chamado in chamados:
    
    if chamado["prazo"] == "Dentro do prazo":
        in_prazo += 1
    else:
        out_prazo += 1
        
        
    if chamado["prioridade"] == "Alta":
        quant_alto += 1
    elif chamado["prioridade"] == "Média":
        quant_medio += 1
    else:
        quant_baixo += 1
        
        

print(f"No total foram realizados {len(chamados)} chamados")
print(f"Há {in_prazo} chamados dentro do prazo")
print(f"Há {out_prazo} chamados fora do prazo")
print(f"O código do chamado que teve o maior tempo de atendimento foi: {chamado_maior_temp["codigo"]}")
print(f"O tempo do chamado dele foi de: {chamado_maior_temp["tempo_chamado_d"]} dias")
print(f"Foram registrados {quant_alto} chamados de alta, {quant_medio} de média e {quant_baixo} de baixa prioridade")
```

## Parte 3

```python
chamados_por_prioridade = {
"Alta": [],
"Média": [],
"Baixa": []
}
```

```python
for chamado in chamados:
    if chamado["prioridade"] == "Alta":
        chamados_por_prioridade["Alta"].append(chamado["codigo"])
    elif chamado["prioridade"] == "Média":
        chamados_por_prioridade["Média"].append(chamado["codigo"])
    else:
        chamados_por_prioridade["Baixa"].append(chamado["codigo"])

chamados_por_prioridade
```

## Desafio

```python
fora_do_prazo = []

for chamado in chamados:
    if chamado["prazo"] == "Fora do prazo":
        dicionario = {}
        dicionario["codigo"] = chamado["codigo"]
        dicionario["responsavel"] = chamado["responsavel"]
        dicionario["tempo"] = chamado["tempo_chamado_d"]
        
        fora_do_prazo.append(dicionario)
    
fora_do_prazo
```

# Atividade Prática 2

```python
transacoes = [
    {
        "codigo": "TR001",
        "cliente": "Maria",
        "tipo": "PIX",
        "valor": 1250.00,
        "horario": "08:15",
        "processamento": 2
    },
    {
        "codigo": "TR002",
        "cliente": "Carlos",
        "tipo": "Transferência",
        "valor": 3500.00,
        "horario": "09:40",
        "processamento": 8
    },
    {
        "codigo": "TR003",
        "cliente": "Marina",
        "tipo": "Pagamento",
        "valor": 850.00,
        "horario": "10:25",
        "processamento": 5
    },
    {
        "codigo": "TR004",
        "cliente": "Rafael",
        "tipo": "PIX",
        "valor": 1250.00,
        "horario": "11:50",
        "processamento": 3
    },
    {
        "codigo": "TR005",
        "cliente": "Beatriz",
        "tipo": "TED",
        "valor": 4200.00,
        "horario": "13:35",
        "processamento": 15
    },
    {
        "codigo": "TR006",
        "cliente": "Lucas",
        "tipo": "PIX",
        "valor": 2100.00,
        "horario": "14:20",
        "processamento": 2
    },
    {
        "codigo": "TR007",
        "cliente": "Fernanda",
        "tipo": "Pagamento",
        "valor": 850.00,
        "horario": "15:45",
        "processamento": 6
    }
]
```

## Parte 1

```python
from datetime import datetime, timedelta 

for numero, transacao in enumerate(transacoes, start=1): 
    try:
        transacao["horario"] = datetime.strptime(transacao["horario"], "%H:%M") 
        
        minutos = int(transacao["processamento"])
        
        tempo_processamento = timedelta(minutes=minutos)
        
        transacao["horario_conclusao"] = transacao["horario"] + tempo_processamento 
            
    except:
        print(f"{numero} - {transacao['codigo']} | {transacao['cliente']} | {transacao['tipo']} | R$ {transacao['valor']} | Inicio: {transacao['horario'].strftime("%H:%M")} | Término: {transacao['horario_conclusao'].strftime("%H:%M")}")        
        print(50*"-")
```

## Parte 2

```python
import statistics
valores = []
```

```python
for transacao in transacoes:
    valores.append(transacao['valor'])

valores
```

```python
def analise_valor(lista):
    valor_medio = statistics.mean(lista)
    valor_mediano = statistics.median(lista)
    moda_valor = statistics.multimode(lista)
    variancia = statistics.pvariance(lista)
    desvio = statistics.pstdev(lista)
    maximo = max(lista)
    minimo = min(lista)
    
    return {"valor_medio": f"{valor_medio:.2f}", 
            "valor_mediano": f"{valor_mediano:.2f}", 
            "moda": f"{moda_valor} (Bimodal)" if len(moda_valor) == 2 else f"{moda_valor} (Multimodal)" if len(moda_valor) > 2 else f"{moda_valor} (Unimodal)", 
            "variancia" : f"{variancia:.2f}",
            "desvio" : f"{desvio:.2f}",
            "max": maximo,
            "minimo": minimo
            }
```

```python
analise_valores = analise_valor(valores)

analise_valores
```

## Parte 3

```python
for numero, transacao in enumerate(transacoes, start=1): 
    valor_medio = float(analise_valores["valor_medio"])
        
    if transacao["valor"] > valor_medio:
        transacao["media"] = "Acima da média"
    elif transacao["valor"] < valor_medio:
        transacao["media"] = "Abaixo da média"
    else:
        transacao["media"] = "Na média"
    
    print(f"{numero} - {transacao['codigo']} | {transacao['cliente']} | R$ {transacao['valor']} | {transacao['media']}")        
    print(50*"-")
        
```

## Parte 4

```python
total = 0
maior_tran = max(transacoes, key=lambda x: x["valor"])
menor_tran = min(transacoes, key=lambda x: x["valor"])

quant_acima = 0
quant_abaixo = 0
quant_medi = 0

for transacao in transacoes:
    
    total += transacao['valor']
    
    if transacao['media'] == "Acima da média":
        quant_acima += 1
    elif transacao['media'] == "Abaixo da média":
        quant_abaixo += 1
    else:
        quant_medi += 1
        
        

print(f"No total foram realizados {len(transacoes)} transações")
print(f"Valor total movimentado: R$ {total}")
print(f"O código da transação que teve o maior valor foi: {maior_tran["codigo"]}")
print(f"O código da transação que teve o menor valor foi: {menor_tran["codigo"]}")
print(f"Foram registrados {quant_acima} transações acima da média e {quant_abaixo} abaixo da média")
```
