### Medidas Centrais 

```python
def calcular_media(lista):
    media = sum(lista)/ len(lista)
    return media
```

```python
def calcular_mediana(lista):
    quantidade = len(lista)
    lista_ordenada = sorted(lista)
    meio = quantidade // 2

    if quantidade % 2 == 0:
        return((lista_ordenada[meio] + lista_ordenada[meio -1]) / 2)
    else:
        return lista_ordenada[meio]
```

```python
print(calcular_mediana([23,14,31,17,20]))
print(calcular_mediana([75,1,4,3,19,50,64,31]))
```

```python
def calcular_modas(valores):
    modas = []
    maior_frequencia = 0

    for valor in valores:
        frequencia = valores.count(valor)

        if frequencia > maior_frequencia:
            maior_frequencia = frequencia
            modas = [valor]

        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)

    if maior_frequencia == 1:
        return "Amodal", []
    elif len(modas) == 1:
        return "Unimodal", modas
    elif len(modas) == 2:
        return "Bimodal", modas
    else:
        return "Multimodal", modas
```

```python
conjunto_a = [ 0, 0, 30, 40]
conjunto_b = [10, 10, 20, 20, 30, 40]
conjunto_c = [10, 10, 20, 20, 30, 30, 40]
conjunto_d = [10, 20, 30, 40]

print(calcular_modas(conjunto_a))
print(calcular_modas(conjunto_b))
print(calcular_modas(conjunto_c))
print(calcular_modas(conjunto_d))
```

### Medidas de Dispersão

```python
def calcular_amplitude(lista):
    amplitude = max(lista) - min(lista)
    return amplitude
```

```python
def calcular_variancia(lista):
    qnt_lista = len(lista)
    media = calcular_media(lista)
    somatoria = 0

    for elemento in lista:
        diferenca_quadrado = pow((elemento - media),2)
        somatoria += diferenca_quadrado
        
    variancia = somatoria / qnt_lista

    return variancia
```

```python
import math 
def calcular_desvio_padrao(lista):
    variancia = calcular_variancia(lista)
    desvio_padrao = math.sqrt(variancia)

    return desvio_padrao
```

```python
lista = [12, 18, 15, 21, 14, 16]
amplitude= calcular_amplitude(lista)
variancia= calcular_variancia(lista)
desvio= calcular_desvio_padrao(lista)

print("----------Medidas de Dispersão----------")
print("O conjunto de dados é:", lista)
print(f"\nAmplitude: {amplitude:.2f}\nVariância: {variancia:.2f}\nDesvio padrão: {desvio:.2f}")
```

```python
def calcular_quartis(lista):
    qnt_valores = len(lista)
    lista_ordenada = sorted(lista)

    if qnt_valores % 2 != 0:
        posicao_q1 = (qnt_valores + 1) / 4
        posicao_q2 = (qnt_valores + 1) / 2
        posicao_q3 = (3 *(qnt_valores + 1)) / 4

        quartil_1 = lista_ordenada[int(posicao_q1) -1]
        quartil_2 = lista_ordenada[int(posicao_q2) -1]
        quartil_3 = lista_ordenada[int(posicao_q3) -1] 

        return quartil_1, quartil_2, quartil_3
    else:
        #------------ Quartil 1 ---------------------------------
        posicao_q1 = (qnt_valores + 1) / 4
        div_inteira_1 = int(posicao_q1)
        div_decimal_1 = posicao_q1 - div_inteira_1

        diferenca_quartil_1 = lista_ordenada[div_inteira_1] - lista_ordenada[div_inteira_1 - 1]
        proporcao_1 = diferenca_quartil_1 * div_decimal_1
        quartil_1 = lista_ordenada[div_inteira_1 -1] + proporcao_1

        #------------ Quartil 2 ---------------------------------
        posicao_q2 = (qnt_valores + 1) / 2
        div_inteira_2 = int(posicao_q2) 
        div_decimal_2 = posicao_q2 - div_inteira_2

        diferenca_quartil_2 = lista_ordenada[div_inteira_2] - lista_ordenada[div_inteira_2 - 1]
        proporcao_2 = diferenca_quartil_2 * div_decimal_2
        quartil_2 = lista_ordenada[div_inteira_2 -1] + proporcao_2

        #------------ Quartil 3 ---------------------------------
        posicao_q3 = (3 *(qnt_valores + 1)) / 4
        div_inteira_3 = int(posicao_q3) 
        div_decimal_3 = posicao_q3 - div_inteira_3

        diferenca_quartil_3 = lista_ordenada[div_inteira_3] - lista_ordenada[div_inteira_3 - 1]
        proporcao_3 = diferenca_quartil_3 * div_decimal_3
        quartil_3 = lista_ordenada[div_inteira_3 -1] + proporcao_3
        

        return quartil_1, quartil_2, quartil_3
```

```python
def calcular_quartis_otimizado(lista):

    n = len(lista)
    lista_ordenada = sorted(lista)

    posicoes = [
        (n + 1) / 4,
        (n + 1) / 2,
        (3 * (n + 1)) / 4
    ]

    quartis = []

    for posicao in posicoes:
        posicao_inteira = int(posicao)

        if posicao == posicao_inteira:
            valor_quartil = lista_ordenada[posicao_inteira - 1]

        else:
            parte_decimal = posicao - posicao_inteira

            diferenca = lista_ordenada[posicao_inteira] - lista_ordenada[posicao_inteira - 1]
            proporcao = diferenca * parte_decimal
            valor_quartil = lista_ordenada[posicao_inteira - 1] + proporcao

        quartis.append(valor_quartil)

    return quartis
```

```python
def calcular_intervalo_quartilico(lista):
    q1, q2, q3 = calcular_quartis_otimizado(lista)
    intervalo = q3 - q1

    return intervalo
```

```python
vendas = [42, 18, 35, 27, 50, 31, 22]
q1, q2, q3 = calcular_quartis_otimizado(vendas)

print("------Medidas de Dispersão- QUARTIS------")
print("Conjunto de dados:", vendas)
print(f"\nQuartil 1: {q1}\nQuartil 2: {q2}\nQuartil 3: {q3}")
print(f"Intervalo Quartilico: {calcular_intervalo_quartilico(vendas)}")
```

```python
def calcular_cof_variacao(lista):
    media = calcular_media(lista)
    dp = calcular_desvio_padrao(lista)

    coeficiente_variacao = (dp / media) * 100

    return coeficiente_variacao
```

```python
valores = [98, 101, 100, 99, 102, 100, 101]
print(f"Coeficiente de Varição: {calcular_cof_variacao(valores):.2f}%")
```
