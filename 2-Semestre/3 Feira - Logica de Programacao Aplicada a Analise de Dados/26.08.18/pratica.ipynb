## For

```python
tempos = [32, 48, 27, 55, 41, 29]

num_entrega = 1 # Contador para saber o número da entrega
total_entregas = len(tempos) # Quantidade total de entregas
tempo_total = 0
entregas_atrasadas = 0


for tempo in tempos:
    
    print("Entrega ", num_entrega)
    print(f"Demorou {tempo} minutos para chegar ao local")
    
    if tempo > 40:
        print("Infelizmente a entrega não foi realizada a tempo\n")
        entregas_atrasadas += 1
    else:
        print("Felizmente a entrega foi feita dentro do prazo\n")
        
    tempo_total += tempo # Pega o tempo total
    num_entrega += 1 # Aumenta o indice da entrega
    
    
tempo_medio = tempo_total/total_entregas
porcent_atrasados = 100 * entregas_atrasadas / total_entregas

print(f"O tempo total das entregas foi de {tempo_total} minutos")
print(f"O tempo médio das entregas foi de {tempo_medio:.2f} minutos")
print(f"Há {entregas_atrasadas} entregas atrasadas")
print(f"Esse número corresponde a   {porcent_atrasados:.1f}% das entregas realizadas")

```

## While and For

```python
producoes = []
total_producao = 0

print("Os valores a seguir tem que ser entre 299 e 801!!\n")
valor = float(input("Digite o valor do turno(Para parar de adicionar digite -1):"))

while valor != -1: # Se for digitado -1 ele sai do loop
    
    # Analisa se o valor é valido
    if valor < 300 or valor > 800:
        print("Valor inválido. Os valores a seguir tem que ser entre 299 e 801!!")
    else:
        producoes.append(valor) # se for válido ele adiciona ele
        print("Valor adicionado")
        total_producao += valor

    print("---------------------------------------------------------------")
    print("Adicione outro número:")
    valor = float(input("Digite o valor do turno(Para parar de adicionar digite -1):")) # Adiciona outro valor
    
print("---------------------------------------------------------------")
print("Inserção Concluida")
print("---------------------------------------------------------------\n")

if len(producoes) > 0: # Se há turnos ele verifica
    
    # Meta por turno
    meta_concluida = 0
    meta_nao_concluida = 0
    for i in range(len(producoes)):  
        if producoes[i] >= 500:
            print(f"Turno {i+1}: Meta Atingida")
            meta_concluida += 1
        elif producoes[i] >= 400:
            print(f"Turno {i+1}: Abaixo da meta")
            meta_nao_concluida += 1
        else:
            print(f"Turno {i+1}: Estado critico para bater a meta")
            meta_nao_concluida += 1
        
    #Porcentagem meta
    porcent_meta_concluida = meta_concluida / len(producoes) * 100

    # Relatorio
    print("---------------------------------------------------------------\n")
    print(f"O total da produção foi de {total_producao} unidades")
    print(f"A quantidade de turnos foi de {len(producoes)}")
    print(f"A produção média por turno foi de {total_producao/len(producoes):.2f} unidades")
    print(f"Turnos com produção acima da meta: {meta_concluida}")
    print(f"Turnos com produção abaixo da meta: {meta_nao_concluida}")
    print(f"Porcentagem de turnos com produção acima da meta: {porcent_meta_concluida:.2f}%")
else: # Se não tem, encerra o programa
    print("Não há turnos para analisar") 
```
