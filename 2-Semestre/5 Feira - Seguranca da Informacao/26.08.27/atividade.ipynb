## Atividade Prática Individual – Criptografia com Python

### Contexto
Uma empresa de saúde ocupacional realiza exames periódicos de colaboradores de diferentes organizações. Durante suas operações, a empresa precisa transmitir informações confidenciais para empresas parceiras responsáveis pelo processamento e armazenamento de determinados dados. Entre as informações transmitidas podem estar: nome; CPF, empresa que trabalha, resultado do exame. Essas informações não devem ser transmitidas de forma legível para terceiros não autorizados. Nesta atividade, vocês deverão implementar em Python duas formas de proteção das informações: utilizando criptografia simétrica e criptografia assimétrica.




### Parte 1: Criptografia Simétrica
A empresa de saúde ocupacional envia informações diariamente para uma empresa parceira com a qual já mantém uma comunicação estabelecida. As duas empresas já possuem previamente a mesma chave secreta, armazenada de maneira segura.

Desenvolva um programa em Python que:
1. Crie uma informação contendo pelo menos quatro dados confidenciais;
2. Gere uma chave secreta;
3. Exiba a informação original;
4. Utilize a chave secreta para realizar a cifragem;
5. Exiba a informação cifrada;
6. Utilize a mesma chave para realizar a decifragem;
7. Exiba a informação recuperada.

```python
from cryptography.fernet import Fernet
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import hashes
```

```python
dados_colaborador = "Jõao da Silva Pereira, 123.456.789-10, Embraer, Cancêr no estômago"
```

```python
# Simétrica
chave = Fernet.generate_key() # Gera uma chave simétrica
cifra = Fernet(chave) # Ação de cifrar e decifrar
```

```python
print("Informação original: ", dados_colaborador) # Dados originais do colaborador

print("-----------------------------------------------------------------")

dados_cifrados = cifra.encrypt(dados_colaborador.encode()) # Cifra a informação do colaborador, transformando em bytes
print("Informação cifrada: ", dados_cifrados)

print("-----------------------------------------------------------------")

dados_decifrados = cifra.decrypt(dados_cifrados) # Decifra a informação do colaborador, transformando em bytes
print("Informação decifrada: ", dados_decifrados.decode())
```

### Parte 2: Criptografia Assimétrica
Agora, a empresa de saúde ocupacional precisa transmitir uma informação confidencial para umanova empresa parceira.Nesse caso, as duas empresas não possuem previamente uma  mesma chave secreta compartilhada. A empresa que receberá as informações possui seu próprio par de chaves: pública e privada.

Utilizando os mesmos dados da Parte 1, desenvolva uma segunda implementação em Python
que:
1. Gere um par de chaves: pública e privada;
2. Exiba a informação original;
3. Utilize a chave pública da empresa destinatária para realizar a cifragem;
4. Exiba a informação cifrada;
5. Utilize a chave privada correspondente para realizar a decifragem;
6. Exiba a informação recuperada.


```python
# Assimétrica
chave_privada_gama = rsa.generate_private_key( # Gera uma chave privada
    public_exponent=65537,
    key_size=2048
)

chave_publica_gama = chave_privada_gama.public_key() # Gera uma chave pública a partir da chave privada
```

```python
print("Informação original: ", dados_colaborador) # Dados originais do colaborador

print("-----------------------------------------------------------------")

dados_cifrados = chave_publica_gama.encrypt( # Cifra a informação do colaborador, transformando em bytes
    dados_colaborador.encode(),
    padding.OAEP(
        mgf=padding.MGF1(hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)
print("Informação cifrada: ", dados_cifrados)

print("-----------------------------------------------------------------")

dados_decifrados = chave_privada_gama.decrypt( # Decifra a informação do colaborador, transformando em bytes
    dados_cifrados,
    padding.OAEP(
        mgf=padding.MGF1(hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)
print("Informação decifrada: ", dados_decifrados.decode()) # O decode permanece para transformar os bytes em string, para que seja possível exibir a informação decifrada
```

## Parte 3: Comparação

1. **Qual é a principal diferença entre a criptografia simétrica e a criptografia assimétrica em relação às chaves utilizadas?**
   * **R:** A criptografia simétrica possui apenas uma chave, que serve tanto para cifrar quanto para decifrar a mensagem ou os dados. Já a criptografia assimétrica funciona com um par de chaves (uma privada e uma pública). A chave privada permanece sempre em segredo com o seu proprietário, enquanto a chave pública pode ser compartilhada abertamente. A chave pública é utilizada para cifrar os dados e apenas a chave privada correspondente consegue reverter ou decifrar essas informações.

2. **Considerando os dois cenários apresentados, explique por que foram utilizadas abordagens criptográficas diferentes.**
   * **R:** Na primeira parte, utilizou-se a criptografia simétrica por se tratar de uma relação pré-existente de confiança entre as empresas. Como ambas já compartilhavam previamente a chave secreta de forma protegida, essa abordagem tornou-se viável e eficiente. Na segunda parte, como a informação seria enviada para uma nova empresa parceira sem um canal prévio para o envio seguro de uma chave secreta, adotou-se a criptografia assimétrica. Dessa forma, a nova parceira forneceu sua chave pública para cifrar os dados, garantindo que apenas ela, com sua chave privada, pudesse lê-los.
