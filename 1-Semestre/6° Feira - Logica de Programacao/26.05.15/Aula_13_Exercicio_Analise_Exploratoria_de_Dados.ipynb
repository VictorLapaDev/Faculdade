#### 8) Qual o pico de novos casos diários no Brasil e quando ocorreu?

#### 9) Crie um gráfico com os novos casos registrados diariamente no Brasil

```python
import matplotlib.pyplot as plt

br = df_brasil.set_index('date')
plt.figure(figsize=(12,8))

plt.plot(br.index,br['new_cases'])
#plt.bar(br.index,br['new_cases'])


plt.title("Casos de covid19 no Brasil")
plt.xlabel("Data")
plt.ylabel("Numero de casos")
plt.xticks([])
plt.show()
```

```python
df_brasil = df_brasil.copy()
df_brasil['date'] = pd.to_datetime(df_brasil['date'])
df_brasil['year'] = df_brasil['date'].dt.year

anos = df_brasil['year'].unique()

fig, axes = plt.subplots(len(anos), 1, figsize=(12,4*len(anos)))

if len(anos) == 1:
    axes = [axes]

for ax, ano in zip(axes,anos):
    dados_ano = df_brasil[df_brasil['year'] == ano]
    ax.plot(dados_ano['date'],dados_ano['new_cases'])
    ax.set_title(f"Cassos de covid19 - {ano}")
    ax.set_ylabel("Casos")
    ax.grid(True)
plt.tight_layout()
plt.show()
```

#### 12) Calcule a taxa de letalidade por país

```python
letalidade = df.copy()
letalidade = letalidade[letalidade['total_cases'] != 0]

letalidade = letalidade.groupby('country').last().reset_index()
letalidade['letalidade'] = letalidade['total_deaths']/letalidade['total_cases']
letalidade = letalidade.sort_values(by='total_cases', ascending=False)

letalidade = letalidade [['country', 'letalidade']].sort_values(by='letalidade', ascending = False)
pd.set_option('display.max_rows', None)
letalidade
```
