```python
def texto_colorido(texto, cor = "branco", estilo = "normal"):
    cores = {
        "vermelho": "\033[31m",
        "verde": "\033[32m",
        "amarelo": "\033[33m",
        "azul": "\033[34m",
        "magenta": "\033[35m",
        "ciano": "\033[36m",
        "branco": "\033[37m"
    }
    
    estilos = {
        "normal": "",
        "negrito": "\033[1m",
        "italico": "\033[3m",
        "sublinhado": "\033[4m"
    }

    if cor in cores and estilo in estilos:
        return f"{estilos[estilo]}{cores[cor]}{texto}\033[0m"
    else:
        return texto  # Retorna o texto sem formatação se a cor ou estilo não forem válidos
```

## Medidas Centrais
São: Média(Todos valores do conjunto), Moda(Maior frequencia), Mediana(Valor central dos dados ordenados)

Modal, multimodal, bimodal, amodal

```python
def calcular_media(lista):
    media = sum(lista)/ len(lista)
    return media
```

```python
def calcular_mediana(lista):

    lista = sorted(lista)
    tamanho = len(lista)
    meio = tamanho // 2

    if tamanho % 2 == 0: # Par

        elemento_1 = lista[meio]
        elemento_2 = lista[meio-1]

        mediana = (elemento_1 + elemento_2) / 2

        return mediana
        
    else: # Impar     
        return lista[meio] 
```

```python
medianaA = calcular_mediana([23, 14, 31, 17, 20]) # 14, 17, 20, 23, 31   20
print(medianaA)

print("-------------------")

medianaB = calcular_mediana([75, 1, 4, 3, 19, 50, 64, 31])
print(medianaB)
```

```python
def calcular_moda(lista):
    moda = None
    frequencia = 0
    for i in lista:
        if lista.count(i) > frequencia:
            frequencia = lista.count(i)
            moda = i
    return moda
```

```python
modaA = calcular_moda([23, 14, 31, 17, 23, 20, 23, 31])
print(modaA)
```

```python
def calcular_moda(lista):
    modas = []
    maior_frequencia = 0

    for valor in lista:
        frequencia = lista.count(valor)

        if frequencia > maior_frequencia:
            maior_frequencia = frequencia
            modas = [valor]

        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)


    if maior_frequencia == 1:
        return "Amodal"
    elif len(modas) == 1:
        return "Unimodal", modas
    elif len(modas) == 2:
        return "Bimodal", modas
    else:
        return "Multimodal", modas
```

# Medidas de Dispersão - 08/09

### Variância

O quanto os dados estão espalhados ao quadrado.

**IMPORTANTE:** Não comparar aos dados originais 

![Variância](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRY7lJblIt7zhjvKF-y_s8vpoKAzlZqgi-NgPswhvZjJYGL0f-KXfg5oJQ&s=10)
### Desvio Padrão 
Medida de dispersão que indica o quanto os valores de um conjunto de dados estão dispersos em relação à média.

**IMPORTANTE:** A sua vantagem é que ele retorna a variância para a mesma unidade dos dados originais 
### Amplitude

```python
def calcular_amplitude(lista):
    amplitude = max(lista) - min(lista)
    return amplitude
```

```python
def calcular_variancia(lista):
    qnt_lista = len(lista)
    media = calcular_media(lista)
    somatoria = 0

    for elemento in lista:
        diferenca_quadrado = pow((elemento - media),2)
        somatoria += diferenca_quadrado
        
    variancia = somatoria / qnt_lista

    return variancia
```

```python
import math 
def calcular_desvio_padrao(lista):
    variancia = calcular_variancia(lista)
    desvio_padrao = math.sqrt(variancia)

    return desvio_padrao
```

```python
def calcular_quartis(lista):
    qnt_valores = len(lista)
    lista_ordenada = sorted(lista)

    if qnt_valores % 2 != 0:
        posicao_q1 = (qnt_valores + 1) / 4
        posicao_q2 = (qnt_valores + 1) / 2
        posicao_q3 = (3 *(qnt_valores + 1)) / 4

        quartil_1 = lista_ordenada[int(posicao_q1) -1]
        quartil_2 = lista_ordenada[int(posicao_q2) -1]
        quartil_3 = lista_ordenada[int(posicao_q3) -1] 

        return quartil_1, quartil_2, quartil_3
    else:
        #------------ Quartil 1 ---------------------------------
        posicao_q1 = (qnt_valores + 1) / 4
        div_inteira_1 = int(posicao_q1)
        div_decimal_1 = posicao_q1 - div_inteira_1

        diferenca_quartil_1 = lista_ordenada[div_inteira_1] - lista_ordenada[div_inteira_1 - 1]
        proporcao_1 = diferenca_quartil_1 * div_decimal_1
        quartil_1 = lista_ordenada[div_inteira_1 -1] + proporcao_1

        #------------ Quartil 2 ---------------------------------
        posicao_q2 = (qnt_valores + 1) / 2
        div_inteira_2 = int(posicao_q2) 
        div_decimal_2 = posicao_q2 - div_inteira_2

        diferenca_quartil_2 = lista_ordenada[div_inteira_2] - lista_ordenada[div_inteira_2 - 1]
        proporcao_2 = diferenca_quartil_2 * div_decimal_2
        quartil_2 = lista_ordenada[div_inteira_2 -1] + proporcao_2

        #------------ Quartil 3 ---------------------------------
        posicao_q3 = (3 *(qnt_valores + 1)) / 4
        div_inteira_3 = int(posicao_q3) 
        div_decimal_3 = posicao_q3 - div_inteira_3

        diferenca_quartil_3 = lista_ordenada[div_inteira_3] - lista_ordenada[div_inteira_3 - 1]
        proporcao_3 = diferenca_quartil_3 * div_decimal_3
        quartil_3 = lista_ordenada[div_inteira_3 -1] + proporcao_3
        

        return quartil_1, quartil_2, quartil_3
```

```python
def calcular_quartis_otimizado(lista):

    n = len(lista)
    lista_ordenada = sorted(lista)

    posicoes = [
        (n + 1) / 4,
        (n + 1) / 2,
        (3 * (n + 1)) / 4
    ]

    quartis = []

    for posicao in posicoes:
        posicao_inteira = int(posicao)

        if posicao == posicao_inteira:
            valor_quartil = lista_ordenada[posicao_inteira - 1]

        else:
            parte_decimal = posicao - posicao_inteira

            diferenca = lista_ordenada[posicao_inteira] - lista_ordenada[posicao_inteira - 1]
            proporcao = diferenca * parte_decimal
            valor_quartil = lista_ordenada[posicao_inteira - 1] + proporcao

        quartis.append(valor_quartil)

    return quartis
```

```python
def calcular_intervalo_quartilico(lista):
    q1, q2, q3 = calcular_quartis_otimizado(lista)
    intervalo = q3 - q1

    return intervalo
```

```python
def realizar_análise(lista):
    media = statistics.mean(lista)
    variancia = statistics.pvariance(lista) # Populacional
    desvio = statistics.pstdev(lista) # Populacional
    amplitude = max(lista) - min(lista)
    cv = (desvio / media) * 100
    return {"Media": media, 
            "Variancia": variancia, 
            "Desvio": desvio, 
            "Amplitude": amplitude,
            "CV(%)": f"{cv:.2f}"}
```

```python
def analisa_vendas(vendedor, *vendas):
    soma_vendas = 0
    for i in vendas:
        soma_vendas += i
    
    media_vendas = soma_vendas/len(vendas)
    
    maior_venda = max(vendas)
    
    if soma_vendas >= 5000:
        meta = "Meta atingida"
    else:
        meta = "Meta não atingida"
    
    print(f"Vendedor: {vendedor}")
    print(f"Valor total vendido: R$ {soma_vendas}")
    print(f"Média das vendas: {media_vendas}")
    print(f"Maior valor de venda: {maior_venda}")
    print(meta)
    print(50*"-")
```

```python
analisa_vendas("José", 500, 600, 500, 2000, 1500, 100, 900, 400)
analisa_vendas("Pedro", 600, 600, 500, 1960, 1500)
analisa_vendas("Lucas", 500, 600, 500, 1400)
```

## Problema 2 - Registro de atendimento

```python
def registrar_atendimento(nome, *procedimentos, **informacoes):
    
    print(f"Nome do cliente: {nome}") # Nome do cliente
    print(f"Quantidade de procedimentos: {len(procedimentos)}") # Quantidade de procedimentos
    
    print(50*"-")
    
    for numero, procedimento in enumerate(procedimentos, start=1): # Mostra os procedimentos
        print(f"{numero}° Procedimento: {procedimento}")
        
    print(50*"-")
    
    if informacoes != {}: # Verifica se há informações adicionais
        print("Informações adicionais:") # Se tiver
        for chave, valor in informacoes.items():
            if chave != "retorno": 
                print(f"{chave}: {valor}")
    else: # Se não tiver
        print("Sem informações adicionais")

    print(50*"-")

    if "retorno" in informacoes: # Se tiver mostra ou diz que não tem data de retorno
        print(f"Data de retorno: {informacoes['retorno']}")
    else:
        print("Retorno não informado")
        
```

```python
registrar_atendimento("Rodrigo", "Mastodonte")
```

## Prática - Cadastro funcionário

```python
def cadastrar_funcionario(nome, idade, salario, setor):
    
    if not nome:
        print("O nome é obrigatório!")
        return

    if idade < 18 or idade > 70:
        print("Idade inválida")
        return

    if salario < 0:
        print("Salário inválido")
        return
    
    if setor not in ["TI", "RH", "Financeiro", "Comercial"]:
        print("Setor inválido")
        return
    
    print("Funcionário cadastrado com sucesso")
```

## Exceção != Validação


Validação = Regra de Negócio

Exceção = Erro de código, sintaxe

```python
try:
    saque = float(input("Insira o valor do seu saque:"))
    parcelas = int(input("Insira o número de parcelas que quer pagar (1 - 6):"))
    
    parcelas_validas = [x for x in range(1,7)]
    
    if saque < 100 or saque > 5000:
        print("Seu saque tem que ser entre R$ 100,00 e R$ 5.000,00")
    
    elif 8000 - saque < 1500:
        print("Seu limite após a operação tem que ser de no minímo de R$ 500,00")
    
    elif parcelas not in parcelas_validas:
        print("Escolha um número de parcelas entre 1 e 6")
    
    else:
        valor_parcela = saque / parcelas
        
        if valor_parcela < 50:
            print("O valor minímo da parcela é de R$ 50,00")
        
        else:
            print("Saldo anterior: R$ 8.000,00")
            print(f"Saque: R$ {saque:.2f}")
            print(f"Saldo restante: R$ {(8000 - saque):.2f}")
            print(f"Número de parcelas: {parcelas}")
            print(f"Valor por parcela: {valor_parcela:.2f}")
        
                
except ValueError:
    print("Insira um valor válido")
except ZeroDivisionError:
    print("Escolha um número de parcelas entre 1 e 6")
except TypeError:
    print("Insira um valor válido")
finally:
    print(50*"-")
    print("Operação finalizada!")
```

```python
try:
    saque = float(input("Insira o valor do seu saque: "))
    parcelas = int(input("Insira o número de parcelas que quer pagar (1 - 6): ")) # Quantas parcelas

    juros_opcao = int(input("Você prefere juros compostos ou juros simples? Digite 1 para composto e 2 para simples: "))
    
    # Verifica se esta no escopo
    if juros_opcao not in [1, 2]:
        print("Insira um valor válido para juros!")
        
    # Se estiver
    else:
        
        if saque < 100 or saque > 5000: # Validação do saque
            print("Seu saque tem que ser entre R$ 100,00 e R$ 5.000,00")
            
        elif 8000 - saque < 1500: # Validação do limite após saque
            print("Seu limite após a operação tem que ser de no mínimo de R$ 500,00")
            
        elif parcelas not in range(1, 7): # Verifica se esta no escopo de parcelas
            print("Escolha um número de parcelas entre 1 e 6")
            
            
            
        else:  # Se passa em todas as regras, calcula o juros

            taxa = 0.02
            
            # Composto
            if juros_opcao == 1:
                
                juros_compostos = lambda s, t, p: s*((1+t) ** p)
                
                taxa = float(input("Digite a taxa de juros em decimais(Ex: 0.02):"))
                if taxa >= 1:
                    taxa = taxa / 100
                    
                valor_pago = juros_compostos(saque, taxa, parcelas)
                
            # Simples
            else: 
                juros_simples = lambda s, p: (s*0.02*p) + s
                valor_pago = juros_simples(saque, parcelas)
            
            
            valor_parcela = valor_pago / parcelas 
            
        
            if valor_parcela < 50: # Validação se a parcela é pelo menos 50 R$
                print("O valor mínimo da parcela é de R$ 50,00")
            else:
                print("\n" + 50*"=")
                print(f"Valor com {'juros compostos' if juros_opcao == 1 else 'juros simples'}")
                print("Saldo anterior: R$ 8.000,00")
                print(f"Saldo restante: R$ {(8000 - valor_pago):.2f}")
                print(f"Saque: R$ {saque:.2f}")
                print(f"Valor a ser pago com juros: R$ {valor_pago:.2f}")
                print(f"Juros: R${valor_pago-saque:.2f}")
                print(f"Taxa de juros: {taxa*100}%")
                print(f"Número de parcelas: {parcelas}")
                print(f"Valor por parcela: R$ {valor_parcela:.2f}")
                print(50*"=")
                
except ValueError:
    print("Insira um valor válido")
except ZeroDivisionError:
    print("Escolha um número de parcelas entre 1 e 6")
except TypeError:
    print("Insira um valor válido")
finally:
    print("Operação finalizada!")
```

```python

def registrar_notas(aluno, *notas):
    
    print(50*"=")

    if not isinstance(aluno, str) or not aluno:
        return print("Insira um aluno válido")
    
    if len(notas) < 1:
        return print("Insira pelo menos uma nota")
    
    for nota in notas:
        if not isinstance(nota, (int, float)):
            return print("Insira notas válidas")
        elif nota > 10 or nota < 0:
            return print("Só são aceitas notas entre 0 e 10")
        
    media = lambda l: sum(l)/len(l)
    
    print(f"Aluno: {aluno}")
    print(f"Total de notas: {len(notas)}")
    print(f"Notas: {notas}")
    print(f"Média: {(media(notas)):.2f}")
    
```

```python
registrar_notas("Felipe", 8, 7.5, 9)
registrar_notas("", 8, 7.5)
registrar_notas("Felipe")
registrar_notas("Felipe", 8, "ç")
registrar_notas("Felipe", 8, 100)
```

```python
def cadastrar_funcionario(nome, *habilidades, **dados):
        
    print(50*"=")

    # Valida nome
    if not isinstance(nome, str) or not nome:
        return print("Insira um nome válido")
    
    # Valida habilidades
    if len(habilidades) < 1:
        return print("Por favor insira uma habilidade")
    
    for habilidade in habilidades:
        if not isinstance(habilidade, str) or not habilidade:
            return print("Insira habilidades válidas")
    
    
    # Valida dados
    if "idade" in dados:
        if not isinstance(dados["idade"], int) or dados['idade'] < 18 or dados['idade'] > 70:
            return print("Insira uma idade válida!")
        
    if "setor" in dados:
        if not isinstance(dados["setor"], str) or dados['setor'] not in ["TI", "Dados", "RH", "Financeiro"]:
            return print("Insira um setor válida!")
    
    if "salario" in dados:
        if not isinstance(dados["salario"], (int, float)) or dados['salario'] <= 0:
            return print("Insira um salário válido!")
        
    if "ativo" in dados:
        if not isinstance(dados["ativo"], bool):
            return print("Insira um ativo válido!")
    
    funcionario = {
        "nome": nome,
        "habilidades": list(habilidades),  
        **dados                           
    }
    
    print("Funcionário cadastrado com sucesso!")
    return funcionario    
```

# Desafio - 20/10

```python
def calcular_frete(cliente, *pesos, **informacoes):
    # Valida cliente
    if not isinstance(cliente, str) or not cliente:
        raise ValueError("Insira um nome válido")
    
    # Valida pesos
    if len(pesos) < 1:
        raise ValueError("Por favor insira ao menos um peso")
    
    peso_total = 0
    
    for peso in pesos:
        if not isinstance(peso, (int, float)) or not peso:
            raise TypeError("Insira um peso válido")
        
        if peso <= 0 or peso > 30:
            print("Os pesos devem ser maiores que 0 e não devem ultrapassar 30 kg")
            return
        
        peso_total += peso
        
    
    if peso_total > 100:
        print("A soma dos pesos não pode ultrapassar 100kg")
        return
    
    # Valida informações
    # Regiao
    regioes = [ "Sudeste", "Sul", "Centro-Oeste", "Nordeste", "Norte"]
    if "regiao" not in informacoes:
        print("Por favor, insira uma região")
        return
    else:
        if informacoes["regiao"] not in regioes:
            print(f"{informacoes['regiao']} é uma região inválida, escolha entre: {', '.join(regioes)}")
            return
    
    # Modalidade transporte
    modalidades = ["Econômica", "Expressa"]
    if 'modalidade' not in informacoes:
        print("Por favor, insira uma modalidade")
        return
    else:
        if informacoes['modalidade'] not in modalidades:
            print(f"{informacoes['modalidade']} é uma modalidade inválida, escolha entre: {', '.join(modalidades)}")
            return
    
    # Seguro
    if 'seguro' not in informacoes:
        informacoes['seguro'] = False
    else:
        if not isinstance(informacoes['seguro'], bool):
            raise ValueError(f"{informacoes['seguro']} é uma opção inválida, por favor escolha entre: True ou False")
        
    
    
    # Calculos
    tarifa_regiao = {
        "Sudeste": 5,
        "Sul": 6,
        "Centro-Oeste": 7,
        "Nordeste": 8,
        "Norte": 10
    }
    
    # Frete base
    frete_base = peso_total * tarifa_regiao[informacoes['regiao']]
    
    # Frete de modalide
    if informacoes['modalidade'] == "Econômica":
        frete_transporte = frete_base
    else:
        frete_transporte = frete_base * 1.4
    
    # Frete com seguro 
    if informacoes['seguro']:
        frete_final = frete_transporte * 1.05
    else:
        frete_final = frete_transporte
        
    valor_seguro = frete_final - frete_transporte
    
    
    print(f"Cliente: {cliente}")
    print(f"Peso total: {peso_total:.2f} kg")
    print(f"Frete-base: R$ {frete_base:.2f} reais")
    print(f"Frete com modalidade: R$ {frete_transporte:.2f} reais")
    print(f"Seguro: R$ {valor_seguro:.2f} reais")
    print(f"Frete final: R$ {frete_final:.2f} reais")
    print("\033[92mEncomenda aprovada!\033[0m")
    print("")
```
