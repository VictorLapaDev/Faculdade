# Classe pronta

```python
class Musica:
    def __init__(self, titulo: str, artista: str, duracao: int, genero: str):
        self.titulo = titulo
        self.artista = artista
        self.duracao = duracao
        self.genero = genero
        
        # Atributos de Controle
        self.reproducoes = 0
        self.tocando = False
        
    def reproduzir(self):
        if self.tocando == False:
            print(f"Reproduzindo música: {self.titulo}")
            self.reproducoes += 1
            self.tocando = True
        else:
            print("Esta música já esta sendo reproduzida")
            
    def pausar(self):
        if self.tocando == True:
            print(f"Pausando música: {self.titulo}")
            self.tocando = False
        else:
            print("A música não está sendo reproduzida.")
    
    def alterar_genero(self, genero):
        self.genero = genero
        
    def exibir_informacoes(self):  
        print(f"Música: {self.titulo}")
        print(f"Artista: {self.artista}")
        print(f"Duração: {self.duracao} s")
        print(f"Reproduções: {self.reproducoes}")
        print(f"Gênero: {self.genero}")
```

```python
m1 = Musica("Vida Boa", "Vitor & Leo", 120, "Sertanejo")
m2 = Musica("Vida Drama", "Vitor & Leo", 124, "Sertanejo")
```

```python
m1.reproduzir()
print(f"Quantidade de reproduções de m1: {m1.reproducoes}")
m1.reproduzir()
m1.pausar()

```

# Criando classe jogos

```python
class Jogos:
    def __init__(self, titulo: str, genero: str, desenvolvedora: str, duracao_campanha: int, preco: float, nota: float):
        self.titulo = titulo
        self.genero = genero
        self.desenvolvedora = desenvolvedora
        self.duracao_campanha = duracao_campanha
        self.preco = preco
        self.nota = nota
        
    def atualizar_preco(self, novo_preco: float):
        self.preco = novo_preco
        print(f"O preço do jogo {self.titulo} foi atualizado para R$ {self.preco:.2f}")
        print("-----------------------------")

    def exibir_informacoes(self):
        print(f"Nome do jogo: {self.titulo}")
        print(f"Gênero: {self.genero}")
        print(f"Desenvolvedora: {self.desenvolvedora}")
        print(f"Duração da campanha: {self.duracao_campanha} dias")
        print(f"Preço: R$ {self.preco:.2f}")
        print(f"Nota: {self.nota}")
        print("-----------------------------")

```

```python
jogo1 = Jogos("The Last of Us", "Ação/Aventura", "Naughty Dog", 15, 199.90, 9.5)
jogo2 = Jogos("God of War", "Ação/Aventura", "Santa Monica Studio", 20, 249.90, 9.8)
jogo3 = Jogos("Horizon Zero Dawn", "Ação/Aventura", "Guerrilla Games", 30, 149.90, 9.2)
```

```python
jogo1.exibir_informacoes()

jogo1.atualizar_preco(179.90)

jogo1.exibir_informacoes()
```
