# Exercício 1

```python
def criar_arvore(valores):
    
    arvore = {}
    
    for valor in valores:
    
        # Verifica se a árvore está vazia (primeiro elemento)
        if len(arvore) == 0:
            # Cria o nó raiz com ponteiros para esquerda e direita vazios
            arvore[valor] = {"esquerda": None, "direita": None}
            raiz = valor  # Define e guarda quem é a raiz principal da árvore
    
        # Se a árvore já tiver elementos
        else:
            # Reinicia a busca sempre a partir da raiz principal
            atual = raiz 
    
            # Loop para descer na árvore até encontrar a posição correta
            while True:
    
                # Se o valor for menor, vai para a esquerda
                if valor < atual:
    
                    # Se o lado esquerdo estiver vazio, insere o novo nó aqui
                    if arvore[atual]["esquerda"] is None: 
                        arvore[atual]["esquerda"] = valor # Conecta o nó atual ao novo valor
                        arvore[valor] = {"esquerda": None, "direita": None}  # Inicializa a estrutura do novo nó
                        break
    
                    # Se já existir um nó na esquerda, desce para ele e continua o loop
                    else:
                        atual = arvore[atual]["esquerda"]
    
                # Se o valor for maior ou igual, vai para a direita
                else:
    
                    # Se o lado direito estiver vazio, insere o novo nó aqui
                    if arvore[atual]["direita"] is None: 
                        arvore[atual]["direita"] = valor # Conecta o nó atual ao novo valor
                        arvore[valor] = {"esquerda": None, "direita": None}  # Inicializa a estrutura do novo nó
                        break
                    
                    # Se já existir um nó na direita, desce para ele e continua o loop
                    else:
                        atual = arvore[atual]["direita"]
    return arvore, raiz
valores = [53, 30, 72, 14, 39, 61, 84, 9, 23, 34, 49, 79]
arvore, raiz = criar_arvore(valores)

print("Árvore criada:")
print(arvore)
print(f"Raiz: {raiz}(primeiro valor do vetor)")
```

# Exercício 2

```python
def inserir_valor_arvore(arvore, valor_inserir):
    

    if len(arvore) == 0:
    
        arvore[valor_inserir] = {
            "esquerda": None,
            "direita": None
        }
    
        raiz = valor_inserir
    
        print(f"O valor {valor_inserir} foi inserido como raiz da árvore.")
    
    else:
        for i in arvore:
            raiz = i
            break
            
        atual = raiz
    
        caminho_insercao = []
    
        while True:
    
            caminho_insercao.append(atual)
    
            #print(f"\nNó atual: {atual}")
    
            if valor_inserir < atual:
    
                #print(f"{valor_inserir} < {atual}")
                #print("Mover para ESQUERDA")
    
                if arvore[atual]["esquerda"] is None:
    
                    arvore[atual]["esquerda"] = valor_inserir
    
                    arvore[valor_inserir] = {
                        "esquerda": None,
                        "direita": None
                    }
    
                    #print(f"Valor {valor_inserir} inserido à ESQUERDA de {atual}.")
                    break
    
                else:
                    atual = arvore[atual]["esquerda"]
    
            else:
    
                #print(f"{valor_inserir} >= {atual}")
                #print("Mover para DIREITA")
    
                if arvore[atual]["direita"] is None:
    
                    arvore[atual]["direita"] = valor_inserir
    
                    arvore[valor_inserir] = {
                        "esquerda": None,
                        "direita": None
                    }
    
                    #print(f"Valor {valor_inserir} inserido à DIREITA de {atual}.")
                    break
    
                else:
                    atual = arvore[atual]["direita"]

    print(f"Valor inserido: {valor_inserir}")
    print(f"\nCaminho da inserção: {caminho_insercao}")
    print(f"\nArvore: \n{arvore}")
    print('\n-----------------------------------')
```

```python
inserir_valor_arvore(arvore, 65)
inserir_valor_arvore(arvore, 90)
inserir_valor_arvore(arvore, 18)
```

# Exercício 3

```python
def procurar_valor_arvore(arvore, valor_procurado):
    
    caminho = []
    quant_nos = 0
    
    for i in arvore:
        raiz = i
        break
        
    atual = raiz # Muda para a raiz da arvore
    
    encontrou = False
    
    while atual is not None:
        # Adiciona ao caminho
        caminho.append(atual)
        quant_nos += 1
    
        # Verifica se o valor já é o procurado
        if valor_procurado==atual:
            encontrou = True
            break
    
        # Verifica se o valor é menor que o nó atual, para descer na esquerda da arvore
        elif valor_procurado < atual:
            atual = arvore[atual]["esquerda"]
            
        # Verifica se o valor é maior que o nó atual, para descer na direita da arvore
        else:
            atual = arvore[atual]["direita"]
    
    if encontrou:
        print("\nValor encontrado")
    else:
        print("\nValor não encontrado")
    
    print(f"Caminho do valor: {caminho}")

    print(f"Nós visitados: {quant_nos}")
    
```

```python
procurar_valor_arvore(arvore, 49)
procurar_valor_arvore(arvore, 84)
procurar_valor_arvore(arvore, 100)
```

# Exercício 22 
### Explique: O que é um nó raiz. O que é um nó folha. O que é altura. O que é profundidade. O que é subárvore.


Uma raiz é o começo da arvore onde o primeiro elemento é inserida.
Um no folha é o filho da raiz, pode ser um elemento maior ou menor que a raiza, indicado para a esquerda ou direita.

A altura é a distância folha até o nó de cima.

A profundidade é a distândia da raiz até o nó de baixo.

Uma subárvore é a arvore menor dentro da arvore maior.
