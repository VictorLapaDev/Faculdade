# Qual equipe teve maior desempenho?

```python
import statistics
```

```python
equipe_A = [42, 38, 45, 41, 39, 44, 38]
equipe_B = [30, 52, 35, 48, 29, 55, 38]
```

### Maior desempenho médio

```python
media_A = statistics.mean(equipe_A)
print(f"Média da equipe A: {media_A}")
media_B = statistics.mean(equipe_B)
print(f"Média da equipe B: {media_B}")
```

### Mediana de cada equipe

```python
mediana_A = statistics.median(equipe_A)
print(f"Mediana da equipe A: {mediana_A}")
mediana_B = statistics.median(equipe_B)
print(f"Mediana da equipe B: {mediana_B}")
```

### Moda

```python
moda_A = statistics.mode(equipe_A)
print(f"Moda da equipe A: {moda_A}")
print(f"A moda da equipe é classificada como: Unimodal")

moda_B = statistics.mode(equipe_B)
print(f"Moda da equipe B: {moda_B}")  # Como não tem valores que se repetem, retorna o primeiro valor da lista
print(f"A moda da equipe é classificada como: Amodal")
```

### Amplitude

```python
amplitude_A = max(equipe_A) - min(equipe_A)
print(f"A amplitude da equipe A é de: {amplitude_A}")
amplitude_B = max(equipe_B) - min(equipe_B)
print(f"A amplitude da equipe B é de: {amplitude_B}")
```

## Qual teve melhor resultado?
Considerando a análise de desempenho, a equipe_A foi melhor. A amplitude dela foi muito menor, tendo pouca variação entre os dados, a mediana de ambas são bem próximas, apesar de na equipe_A ser um pouco maior, a média das duas equipes foram iguais e considerando a moda, a equipe_A tem um valor que se repete mais, mostrando constância. Com base em todos estes dados, a equipe_A teve melhor desempenho principalmente pela constância de dados, tendo pouca variação, ficando com valores próximos um do outro.

# Qual máquina é mais regular?

```python
maquina_A = [98, 101, 100, 99, 102, 100, 101]
maquina_B = [82, 115, 91, 108, 120, 86, 99]
```

```python
def realizar_análise(lista):
    media = statistics.mean(lista)
    variancia = statistics.pvariance(lista) # Populacional
    desvio = statistics.pstdev(lista) # Populacional
    amplitude = max(lista) - min(lista)
    cv = (desvio / media) * 100
    return {"Media": media, 
            "Variancia": variancia, 
            "Desvio": desvio, 
            "Amplitude": amplitude,
            "CV(%)": f"{cv:.2f}"}
```

```python
resul_A = realizar_análise(maquina_A)
resul_B = realizar_análise(maquina_B)
print(f"Resultados da máquina A: {resul_A}")
print(f"Resultados da máquina B: {resul_B}")

```

## Qual máquina apresenta produção mais regular?
Com base na análise, a máquina A representa uma produção mais regular. Como sua amplitude é baixa e o desvio padrão é próximo de 1, a produção desta máquina não é oscilante, ela se baseia em valores muitos próximos, enquanto a máquina B representa uma oscilação bem maior.

Com base no coeficiente de variação, o resultado continua o mesmo, a porcentagem de variação com base na média continua bem baixa na máquina A, batendo 1.3%, apesar da máquina B ter um resultado baixo também, essa variação na máquina A é quase nula.

# Análise dos tempos de entrega

```python
tempos = [48, 32, 55, 41, 37, 62, 29, 45, 51, 35, 70]
```

```python
# Da aula passada
def calcular_quartis_otimizado(lista):

    n = len(lista)
    lista_ordenada = sorted(lista)

    posicoes = [
        (n + 1) / 4,
        (n + 1) / 2,
        (3 * (n + 1)) / 4
    ]

    quartis = []

    for posicao in posicoes:
        posicao_inteira = int(posicao)

        if posicao == posicao_inteira:
            valor_quartil = lista_ordenada[posicao_inteira - 1]

        else:
            parte_decimal = posicao - posicao_inteira

            diferenca = lista_ordenada[posicao_inteira] - lista_ordenada[posicao_inteira - 1]
            proporcao = diferenca * parte_decimal
            valor_quartil = lista_ordenada[posicao_inteira - 1] + proporcao

        quartis.append(valor_quartil)

    return quartis

def calcular_intervalo_quartilico(lista):
    q1, q2, q3 = calcular_quartis_otimizado(lista)
    intervalo = q3 - q1

    return intervalo
```

```python
def analise_tempo(lista):
    tempo_medio = statistics.mean(lista)
    tempo_mediano = statistics.median(lista)
    q = calcular_quartis_otimizado(lista)
    intervalo = calcular_intervalo_quartilico(lista)
    amplitude = max(lista) - min(lista)
    variancia = statistics.variance(lista)
    desvio = statistics.stdev(lista)
    cv = (desvio / tempo_medio) * 100
    
    return {"tempo_medio": f"{tempo_medio:.2f}", 
            "tempo_mediano": f"{tempo_mediano:.2f}", 
            "desvio_padrao": f"{desvio:.2f}", 
            "variancia" : f"{variancia:.2f}",
            "amplitude": f"{amplitude:.2f}",
            "CV(%)": f"{cv:.2f}",
            "Q1": q[0],
            "Q2": q[1],
            "Q3": q[2],
            "intervalo_quartilico": intervalo
            }
```

```python
resultados = analise_tempo(tempos)
print(resultados)
```

Com base nos resultados a afirmação do gestor de, "A maioria das entregas ocorre muito próxima da média", pode ser dita como falsa, uma vez que o desvio padrão é alto, com um intertvalo quartilico de 20 referente a mediana, é uma variação muito grande, sendo assim, não, não é um tempo próximo da média.
