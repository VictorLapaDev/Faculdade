### 1. Solicitar Temperatura

### 2. Par ou Ímpar

```python
numero = int(input("Digite um número:"))

if numero % 2 == 0:
    print(f'Número {numero} é Par')
else:
    print(f'Número {numero} é Ímpar')
```

### *Operadores Relacionais* 


* **Igual a (==):** Verifica se dois valores são iguais.
* **Diferente de (!=):** Verifica se dois valores não são iguais.
* **Maior que (>):** Verifica se o primeiro valor é maior que o segundo.
* **Menor que (<):** Verifica se o primeiro valor é menor que o segundo.
* **Maior ou igual a (>=):** Verifica se o primeiro valor é maior ou igual ao segundo.
* **Menor ou igual a (<=):** Verifica se o primeiro valor é menor ou igual ao segundo

### 3. Verificar saldo

```python
saldo = float(input("Digite o seu saldo atual:"))
valor = float(input("Digit o valor de transação:"))
saldo_suficiente= False 

if saldo >= valor:
    saldo_suficiente= True 
    print(f'Transação autorizada. \n Saldo atual: {saldo-valor:.2f}')
else:
    print("Transação negada: saldo insuficiente!")

print("Saldo suficiente:", saldo_suficiente)
```
