```python
import json
import glob
from datetime import datetime
from pathlib import Path

def merge_notebooks(filenames, output_filename):
    """
    Une múltiplos Jupyter Notebooks em um único arquivo
    """
    merged = {
        "cells": [],
        "metadata": {
            "kernelspec": {
                "display_name": "Python 3",
                "language": "python",
                "name": "python3"
            },
            "language_info": {
                "name": "python",
                "version": "3.8.0"
            }
        },
        "nbformat": 4,
        "nbformat_minor": 4
    }
    
    # Adicionar um cabeçalho
    header_cell = {
        "cell_type": "markdown",
        "metadata": {},
        "source": [
            "# Portfólio Consolidado de Análise de Dados\n",
            "## Contendo todos os notebooks dos alunos\n",
            f"### Gerado em: {datetime.now().strftime('%Y-%m-%d %H:%M')}"
        ]
    }
    merged["cells"].append(header_cell)
    
    # Processar cada notebook
    for filename in filenames:
        # Adicionar um separador com o nome do notebook
        separator_cell = {
            "cell_type": "markdown",
            "metadata": {},
            "source": [f"## {Path(filename).stem.replace('_', ' ').title()}"]
        }
        merged["cells"].append(separator_cell)
        
        # Carregar e adicionar o conteúdo do notebook
        try:
            with open(filename, 'r', encoding='utf-8') as f:
                notebook = json.load(f)
                merged["cells"].extend(notebook["cells"])
        except FileNotFoundError:
            print(f"Aviso: Arquivo {filename} não encontrado. Pulando...")
        except json.JSONDecodeError:
            print(f"Aviso: Erro ao ler {filename}. Possível arquivo corrompido.")
    
    # Salvar o notebook consolidado
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(merged, f, indent=2)
    
    print(f"Notebook consolidado salvo como: {output_filename}")
    return output_filename

# Exemplo de uso:
if __name__ == "__main__":
    # Lista de notebooks para unir (em ordem)
    notebooks = [
        "1_ WEBSCRAPING.ipynb",
        "3_Web_Scraping_BeautifulSoup_Capes.ipynb", 
        "4_Web_Scraping_Selenium_Instalação.ipynb",
        "4_Web_Scraping_Selenium_Intro - Pubmed.ipynb",
        "5_Consumo_de_APIs_parte1.ipynb",
        "5_Consumo_de_APIs_parte2.ipynb",
        "6_SQLite.ipynb",
        "7_WebScraping_SQLite.ipynb",
        "7.1_exercicios.ipynb",
        "8_JOIN.ipynb",
        "9_Forma_Normal_FN.ipynb",
        "10_IntroduçãoMySQL.ipynb",
        "11_Procedures_Functions_Triggers.ipynb",
        "12_MySQL_Grant_Revoke.ipynb",
        "13_MySQL_Views.ipynb",
    ]
    
    # Verificar quais notebooks existem
    existing_notebooks = [nb for nb in notebooks if Path(nb).exists()]
    
    if not existing_notebooks:
        print("Nenhum notebook encontrado. Verifique os nomes dos arquivos.")
    else:
        merge_notebooks(existing_notebooks, "Portfolio_completo.ipynb")
```
