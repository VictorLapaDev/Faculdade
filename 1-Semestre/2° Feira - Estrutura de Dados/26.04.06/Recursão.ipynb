```python
def expoente(base, expo):
    if expo == 0:
        return 1
    elif expo == 1:
        return base
    elif base < 0 or expo < 0:
        return None
    return base * expoente(base, expo-1)
    
```

```python
expoente(5, 5)
```

```python
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    return n * fatorial(n - 1)
```

```python
fatorial(10)
```

```python
def fatorial2(num):
    total = 1
    for i in range(1, num+1):
        total *= i
    return total
```

```python
fatorial2(10)
```
