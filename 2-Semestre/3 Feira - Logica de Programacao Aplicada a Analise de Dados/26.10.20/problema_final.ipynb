```python
def calcular_frete(cliente, *pesos, **informacoes):
    # Valida cliente
    if not isinstance(cliente, str) or not cliente:
        raise ValueError("Insira um nome válido")
    
    # Valida pesos
    if len(pesos) < 1:
        raise ValueError("Por favor insira ao menos um peso")
    
    peso_total = 0
    
    for peso in pesos:
        if not isinstance(peso, (int, float)) or not peso:
            raise TypeError("Insira um peso válido")
        
        if peso <= 0 or peso > 30:
            print("Os pesos devem ser maiores que 0 e não devem ultrapassar 30 kg")
            return
        
        peso_total += peso
        
    
    if peso_total > 100:
        print("A soma dos pesos não pode ultrapassar 100kg")
        return
    
    # Valida informações
    # Regiao
    regioes = [ "Sudeste", "Sul", "Centro-Oeste", "Nordeste", "Norte"]
    if "regiao" not in informacoes:
        print("Por favor, insira uma região")
        return
    else:
        if informacoes["regiao"] not in regioes:
            print(f"{informacoes['regiao']} é uma região inválida, escolha entre: {', '.join(regioes)}")
            return
    
    # Modalidade transporte
    modalidades = ["Econômica", "Expressa"]
    if 'modalidade' not in informacoes:
        print("Por favor, insira uma modalidade")
        return
    else:
        if informacoes['modalidade'] not in modalidades:
            print(f"{informacoes['modalidade']} é uma modalidade inválida, escolha entre: {', '.join(modalidades)}")
            return
    
    # Seguro
    if 'seguro' not in informacoes:
        informacoes['seguro'] = False
    else:
        if not isinstance(informacoes['seguro'], bool):
            raise ValueError(f"{informacoes['seguro']} é uma opção inválida, por favor escolha entre: True ou False")
        
    
    
    # Calculos
    tarifa_regiao = {
        "Sudeste": 5,
        "Sul": 6,
        "Centro-Oeste": 7,
        "Nordeste": 8,
        "Norte": 10
    }
    
    # Frete base
    frete_base = peso_total * tarifa_regiao[informacoes['regiao']]
    
    # Frete de modalide
    if informacoes['modalidade'] == "Econômica":
        frete_transporte = frete_base
    else:
        frete_transporte = frete_base * 1.4
    
    # Frete com seguro 
    if informacoes['seguro']:
        frete_final = frete_transporte * 1.05
    else:
        frete_final = frete_transporte
        
    valor_seguro = frete_final - frete_transporte
    
    
    print(f"Cliente: {cliente}")
    print(f"Peso total: {peso_total:.2f} kg")
    print(f"Frete-base: R$ {frete_base:.2f} reais")
    print(f"Frete com modalidade: R$ {frete_transporte:.2f} reais")
    print(f"Seguro: R$ {valor_seguro:.2f} reais")
    print(f"Frete final: R$ {frete_final:.2f} reais")
    print("\033[92mEncomenda aprovada!\033[0m")
    print("")
```

```python
# Testes obrigatórios
calcular_frete("Lapa", 10, 25, 18, regiao="Sul", modalidade="Expressa", seguro=True) # Dados validos
```
