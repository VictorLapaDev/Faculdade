#### 1. Problema: Desempenho de uma rede de Super-Mercados

```python
faturamento = [18500, 21300, 17800, 24600, 27400]

quant_dias = len(faturamento)
fat_total = sum(faturamento)
fat_media = fat_total / quant_dias
fat_max = max(faturamento)
fat_min = min(faturamento)

print("-----Relatório Final-----\n")
print("Faturamentos de Segunda à Sexta:", faturamento)
print(f'Quantidade de Dias: {quant_dias}\nFaturamento total: {fat_total}\nFaturamento médio diário: {fat_media}\nFaturamento Máximo: {fat_max}\nFaturamento Minímo: {fat_min}')

if fat_media >= 22000:
    print("Status: Bom Desempenho")
elif fat_media >= 19000 and fat_media <= 21999:
    print("Status: Desempenho Regular")
else:
    print("Status: Desempenho Crítico")
```

#### 2. Controle de Velocidade

```python
velocidade = int(input("Digite a velocidade do veículo:"))
limite = 70

if velocidade > limite:
    print("Velocidade ultrapassada! Motorista foi multado")
    valor_multa = (velocidade- limite) * 9
    print("Valor total da Multa:", valor_multa)
    execesso_percentual = ((velocidade - limite) / limite) * 100 
    print(f"Percentual de excesso de velocidade: {execesso_percentual:.2f}%")
    if execesso_percentual > 50:
        print("\nFicará um ano sem sua CNH!!!")
else:
    print("Velocidade dentro do limite")
```

#### 3. Monitoramento de Produção Industrial

```python
producao = [485, 520, 472, 498, 455, 520]

turnos = len(producao)
prod_total = sum(producao)
prod_media = prod_total / turnos
prod_max = max(producao)
prod_min = min(producao)

print("-----Relatório Final-----\n")
print("Produções por turnos:", producao)
print(f'Quantidade de Turnos: {turnos}\nProdução total: {prod_total}\nProdução médio por turno: {prod_media}\nProdução Máxima: {prod_max}\nProdução Minímo: {prod_min}\n')

if prod_media >= 500:
    print("Meta atingida!\n")
elif prod_media >= 480 and prod_media <= 499.99:
    print("Atenção à meta!\n")
else:
    print("Abaixo da meta!\n")

if prod_min < 460:
    print("Alerta operacional!")
```

#### 4. Análise de Rentabilidade de uma Empresa

```python
janeiro = [85000, 46000, 18000]
fevereiro = [92000, 51000, 19000]
marco = [78000, 48000, 20000]
abril = [105000, 54000, 21000]
maio = [88000, 52000, 22000]

result_jan = janeiro[0] - janeiro[1] - janeiro[2]
result_fev = fevereiro[0] - fevereiro[1] - fevereiro[2]
result_mar = marco[0] - marco[1] - marco[2]
result_abr = abril[0] - abril[1] - abril[2]
result_mai = maio[0] - maio[1] - maio[2]

# Janeiro 
imo_jan = (result_jan / janeiro[0]) * 100
if imo_jan >= 25 :
    class_jan= "Excelente"
elif imo_jan >= 15:
    class_jan= "Bom" 
elif imo_jan >= 8:
    class_jan= "Regular"
elif imo_jan > 0:
    class_jan= "Atenção"
elif imo_jan == 0:
    class_jan= "Ponto de equilibrio"
else: 
    class_jan= "Prejuízo"

#Fevereiro
imo_fev = (result_fev / fevereiro[0]) * 100
if imo_fev >= 25 :
    class_fev= "Excelente"
elif imo_fev >= 15:
    class_fev= "Bom" 
elif imo_fev >= 8:
    class_fev= "Regular"
elif imo_fev > 0:
    class_fev= "Atenção"
elif imo_fev == 0:
    class_fev= "Ponto de equilibrio"
else: 
    class_fev= "Prejuízo"

#Março
imo_mar = (result_mar / marco[0]) * 100
if imo_mar >= 25 :
    class_mar= "Excelente"
elif imo_mar >= 15:
    class_mar= "Bom" 
elif imo_mar >= 8:
    class_mar= "Regular"
elif imo_mar > 0:
    class_mar= "Atenção"
elif imo_mar == 0:
    class_mar= "Ponto de equilibrio"
else: 
    class_mar= "Prejuízo"

#Abril
imo_abr = (result_abr / abril[0]) * 100
if imo_abr >= 25 :
    class_abr= "Excelente"
elif imo_abr >= 15:
    class_abr= "Bom" 
elif imo_abr >= 8:
    class_abr= "Regular"
elif imo_abr > 0:
    class_abr= "Atenção"
elif imo_abr == 0:
    class_abr= "Ponto de equilibrio"
else: 
    class_abr= "Prejuízo"

#Maio
imo_mai = (result_mai / maio[0]) * 100
if imo_mai >= 25 :
    class_mai= "Excelente"
elif imo_mai >= 15:
    class_mai= "Bom" 
elif imo_mai >= 8:
    class_mai= "Regular"
elif imo_mai > 0:
    class_mai= "Atenção"
elif imo_mai == 0:
    class_mai= "Ponto de equilibrio"
else: 
    class_mai= "Prejuízo"


lista_imo = [imo_jan, imo_fev, imo_mar, imo_abr, imo_mai]
imo_max = max(lista_imo)
imo_min = min(lista_imo)

if imo_max == imo_jan:
    max_imo_mes = "Janeiro"
elif imo_max == imo_fev:
    max_imo_mes = "Fevereiro"
elif imo_max == imo_mar:
    max_imo_mes = "Março"
elif imo_max == imo_abr:
    max_imo_mes = "Abril"
else:
    max_imo_mes = "Maio"

if imo_min == imo_jan:
    min_imo_mes = "Janeiro"
elif imo_min == imo_fev:
    min_imo_mes = "Fevereiro"
elif imo_min == imo_mar:
    min_imo_mes = "Março"
elif imo_min == imo_abr:
    min_imo_mes = "Abril"
else:
    min_imo_mes = "Maio"

print


print("-----Relatórios Mensais-----")
print("\nJaneiro")
print(f"\tResultado financeiro: {result_jan}\n\tÍndice de Margem Operacional: {imo_jan:.1f}%\n\tClassificação: {class_jan}\n")

print("Fevereiro")
print(f"\tResultado financeiro: {result_fev}\n\tÍndice de Margem Operacional: {imo_fev:.1f}%\n\tClassificação: {class_fev}\n")

print("Março")
print(f"\tResultado financeiro: {result_mar}\n\tÍndice de Margem Operacional: {imo_mar:.1f}%\n\tClassificação: {class_mar}\n")

print("Abril")
print(f"\tResultado financeiro: {result_abr}\n\tÍndice de Margem Operacional: {imo_abr:.1f}%\n\tClassificação: {class_abr}\n")

print("Maio")
print(f"\tResultado financeiro: {result_mai}\n\tÍndice de Margem Operacional: {imo_mai:.1f}%\n\tClassificação: {class_mai}\n")

print("Máximo e Minímo IMO:")
print(f"  Maior Índice de Margem Operacional {imo_max:.2f} obtido no mês {max_imo_mes}")
print(f"  Minímo Índice de Margem Operacional {imo_min:.2f} obtido no mês {min_imo_mes}")
```
