```python
def shell_sort(arr):

    n = len(arr) # Tamanho do vetor
    pulo = n // 2 # Metade do valor, pulo que ele vai fazer
    
    while pulo > 0: # Confere se o pulo é maior que 0 para não percorrer o mesmo número
    
        for i in range(pulo, n): # Percorre do pulo(meio da lista) até o final
            valor = arr[i]   # Pega o valor que esta sendo conferido
            posicao = i # Pega a posicao do valor que esta sendo conferido

            while posicao >= pulo and arr[posicao -pulo] > valor: # Avalia se a posicao é maior que o pulo e se o valor que esta anterior(no pulo) é maior que o da frente
                arr[posicao] = arr[posicao - pulo] # Define o valor da frente igual ao antes do pulo
                posicao -= pulo # Muda a posicao para ficar igual ao do valor que foi jogado para frente

            arr[posicao] = valor # Define o valor que estava na frente como o valor antes do pulo
            # No fim, ele apenas troca as duas variaveis

        pulo //= 2 # Diminui o pulo para executar mais vezes e terminar a lista

    return arr
        
```

```python
array = [2,3,5,6,3,2,3,4,2,3,7,8,9,345,2,234,345,456,456,32,34,46,453,21]

ordenado = shell_sort(array)
print(ordenado)
```
