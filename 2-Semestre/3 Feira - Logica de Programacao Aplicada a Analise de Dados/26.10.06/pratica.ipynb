## Exceção != Validação


Validação = Regra de Negócio

Exceção = Erro de código, sintaxe

```python
try:
    saque = float(input("Insira o valor do seu saque:"))
    parcelas = int(input("Insira o número de parcelas que quer pagar (1 - 6):"))
    
    parcelas_validas = [x for x in range(1,7)]
    
    if saque < 100 or saque > 5000:
        print("Seu saque tem que ser entre R$ 100,00 e R$ 5.000,00")
    
    elif 8000 - saque < 1500:
        print("Seu limite após a operação tem que ser de no minímo de R$ 500,00")
    
    elif parcelas not in parcelas_validas:
        print("Escolha um número de parcelas entre 1 e 6")
    
    else:
        valor_parcela = saque / parcelas
        
        if valor_parcela < 50:
            print("O valor minímo da parcela é de R$ 50,00")
        
        else:
            print("Saldo anterior: R$ 8.000,00")
            print(f"Saque: R$ {saque:.2f}")
            print(f"Saldo restante: R$ {(8000 - saque):.2f}")
            print(f"Número de parcelas: {parcelas}")
            print(f"Valor por parcela: {valor_parcela:.2f}")
        
                
except ValueError:
    print("Insira um valor válido")
except ZeroDivisionError:
    print("Escolha um número de parcelas entre 1 e 6")
except TypeError:
    print("Insira um valor válido")
finally:
    print(50*"-")
    print("Operação finalizada!")
```

```python
try:
    saque = float(input("Insira o valor do seu saque: "))
    parcelas = int(input("Insira o número de parcelas que quer pagar (1 - 6): ")) # Quantas parcelas

    juros_opcao = int(input("Você prefere juros compostos ou juros simples? Digite 1 para composto e 2 para simples: "))
    
    # Verifica se esta no escopo
    if juros_opcao not in [1, 2]:
        print("Insira um valor válido para juros!")
        
    # Se estiver
    else:
        
        if saque < 100 or saque > 5000: # Validação do saque
            print("Seu saque tem que ser entre R$ 100,00 e R$ 5.000,00")
            
        elif 8000 - saque < 1500: # Validação do limite após saque
            print("Seu limite após a operação tem que ser de no mínimo de R$ 500,00")
            
        elif parcelas not in range(1, 7): # Verifica se esta no escopo de parcelas
            print("Escolha um número de parcelas entre 1 e 6")
            
            
            
        else:  # Se passa em todas as regras, calcula o juros

            taxa = 0.02
            
            # Composto
            if juros_opcao == 1:
                
                juros_compostos = lambda s, t, p: s*((1+t) ** p)
                
                taxa = float(input("Digite a taxa de juros em decimais(Ex: 0.02):"))
                if taxa >= 1:
                    taxa = taxa / 100
                    
                valor_pago = juros_compostos(saque, taxa, parcelas)
                
            # Simples
            else: 
                juros_simples = lambda s, p: (s*0.02*p) + s
                valor_pago = juros_simples(saque, parcelas)
            
            
            valor_parcela = valor_pago / parcelas 
            
        
            if valor_parcela < 50: # Validação se a parcela é pelo menos 50 R$
                print("O valor mínimo da parcela é de R$ 50,00")
            else:
                print("\n" + 50*"=")
                print(f"Valor com {'juros compostos' if juros_opcao == 1 else 'juros simples'}")
                print("Saldo anterior: R$ 8.000,00")
                print(f"Saldo restante: R$ {(8000 - valor_pago):.2f}")
                print(f"Saque: R$ {saque:.2f}")
                print(f"Valor a ser pago com juros: R$ {valor_pago:.2f}")
                print(f"Juros: R${valor_pago-saque:.2f}")
                print(f"Taxa de juros: {taxa*100}%")
                print(f"Número de parcelas: {parcelas}")
                print(f"Valor por parcela: R$ {valor_parcela:.2f}")
                print(50*"=")
                
except ValueError:
    print("Insira um valor válido")
except ZeroDivisionError:
    print("Escolha um número de parcelas entre 1 e 6")
except TypeError:
    print("Insira um valor válido")
finally:
    print("Operação finalizada!")
```

```python

def registrar_notas(aluno, *notas):
    
    print(50*"=")

    if not isinstance(aluno, str) or not aluno:
        return print("Insira um aluno válido")
    
    if len(notas) < 1:
        return print("Insira pelo menos uma nota")
    
    for nota in notas:
        if not isinstance(nota, (int, float)):
            return print("Insira notas válidas")
        elif nota > 10 or nota < 0:
            return print("Só são aceitas notas entre 0 e 10")
        
    media = lambda l: sum(l)/len(l)
    
    print(f"Aluno: {aluno}")
    print(f"Total de notas: {len(notas)}")
    print(f"Notas: {notas}")
    print(f"Média: {(media(notas)):.2f}")
    
```

```python
registrar_notas("Felipe", 8, 7.5, 9)
registrar_notas("", 8, 7.5)
registrar_notas("Felipe")
registrar_notas("Felipe", 8, "ç")
registrar_notas("Felipe", 8, 100)
```
