```python
def gerar_escada(num):
    # Percorre cada linha da escada (de 0 até num-1)
    for x in range(num):
        texto = ""  # Inicializa a string que armazenará os caracteres da linha atual
        
        # Percorre as colunas da esquerda para a direita (de 0 até num-1)
        for y in range(num):
            
            # Se o número da coluna (y) for menor ou igual ao número da linha (x),
            # adiciona "*" à string.
            # Caso contrário, adiciona um espaço em branco.
            # Isso faz com que os asteriscos aumentem a cada linha,
            # formando uma escada crescente.
            if y <= x:
                texto += "*"
            else:
                texto += " "
        
        # Exibe a linha completa após montar todos os caracteres
        print(texto)
```

```python
gerar_escada(10)
```

```python
def gerar_escada_contraria(num):
    # Percorre cada linha da escada (de 0 até num-1)
    for x in range(num):
        texto = ""  # Inicializa a string que irá armazenar os caracteres da linha atual
        
        # Percorre as colunas da esquerda para a direita (de 0 até num-1)
        for y in range(num):
            
            # Se o número da coluna (y) for maior ou igual ao número da linha (x),
            # adiciona "*" à string.
            # Caso contrário, adiciona um espaço em branco.
            # Isso faz com que os asteriscos diminuam da esquerda para a direita
            # a cada nova linha, formando uma escada ao contrário.
            if y >= x:
                texto += "*"
            else:
                texto += " "
        
        # Imprime a linha completa depois de montar todos os caracteres
        print(texto)
```

```python
gerar_escada_contraria(9)
```

```python
def gerar_escada_inversa(num):
    # Percorre cada linha da escada (de 0 até num-1)
    for x in range(num):
        texto = ""  # Inicializa a string que irá armazenar os caracteres da linha atual
        
        # Percorre as colunas de trás para frente (de num até 1)
        for y in range(num, 0, -1):
            
            # Se o número da coluna (y) for menor ou igual ao número da linha (x),
            # adiciona um espaço em branco.
            # Caso contrário, adiciona "*".
            # Isso faz com que os asteriscos diminuam a cada linha,
            # formando uma escada invertida.
            if y <= x:
                texto += " "
            else:
                texto += "*"
        
        # Exibe a linha completa após montar todos os caracteres
        print(texto)
```

```python
gerar_escada_inversa(9)
```

```python
def gerar_escada_inversa_contraria(num):
    # Percorre cada linha da escada (de 0 até num-1)
    for x in range(num):
        texto = "" # Inicializa a string que irá armazenar os caracteres da linha atual
        
        # Percorre as colunas da direita para a esquerda (de num até 1)
        for y in range(num, 0, -1):
            
            # Se o valor atual da coluna (y) for menor ou igual ao número da linha (x),
            # adiciona "*" à string. Isso faz com que os asteriscos apareçam
            # progressivamente à direita, formando a escada.
            # Caso contrário, adiciona um espaço em branco.
            if y <= x:
                texto += "*"
            else:
                texto += " "
        # Imprime a linha completa após terminar de montar todos os caracteres
        print(texto)
```

```python
def caixa_pago(valor_compra, valor_pago):
    
    if valor_pago < valor_compra:
        saldo_devedor = valor_compra - valor_pago
        print(f"Falta ser pago {saldo_devedor} R$ pela compra.")
        
    elif valor_pago > valor_compra:
        
        valor_troco = valor_pago - valor_compra
        print(f"Seu troco será de {valor_troco} R$.")
        
        notas = [100, 50, 20, 10, 5, 2, 1]
        for x in notas:
            quant_notas = valor_troco // x #retorna o valor inteiro da divisão
            if quant_notas > 0:
                print(f"{quant_notas} nota(s) de {x} R$")
                valor_troco -= (quant_notas * x) 
                
    else:
        print("O valor da compra foi abatido.")
```

```python
def num_par(num):
    """
    0 é para 0
    1 é para par
    2 é para impar
    """
    if num == 0:
        return 0
    elif num % 2 == 0:
        return 1
    else:
        return 2

def contar_pares_impares(lista):
    
    numero_par = 0
    numero_impar = 0
    numero_invalido = 0

    for i in lista:
    
        option = num_par(i)
    
        if option == 0:
            numero_invalido += 1
        elif option == 1:
            numero_par += 1
        else:
            numero_impar += 1

    return numero_par, numero_impar, numero_invalido
```

```python
lista = [1, 2, 4, 5, 1, 78, 56, 0]
par, impares, invalidos = contar_pares_impares(lista)
print(f"Números pares: {par} | Números impares: {impares} | Números inválidos: {invalidos}")
```
