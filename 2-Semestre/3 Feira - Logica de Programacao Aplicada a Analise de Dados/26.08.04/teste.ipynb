```python
print("Bem vindo a calculadora básica de operações matemáticas!!\n")
print(
    "A seguir temos uma calculadora básica que irá realizar as 7 operações matemáticas básicas com dois números que você irá fornecer.\nAs operações são: soma, subtração, multiplicação, divisão, resto da divisão, exponenciação e divisão inteira."
)
print(
    "Será necessário informar dois números para que a calculadora funcione corretamente.\n"
)

num1 = float(input("Digite o primeiro número:\n\n"))
num2 = float(input("Digite o segundo número:\n\n"))

print(f"A soma entre {num1} e {num2} é: {num1 + num2}\n")
print(f"A subtração entre {num1} e {num2} é: {num1 - num2}\n")
print(f"A multiplicação entre {num1} e {num2} é: {num1 * num2}\n")
print(f"A exponenciação entre {num1} e {num2} é: {num1 ** num2}\n")
print(f"A divisão entre {num1} e {num2} é: {(num1 / num2):.2f}\n")
print(f"O resto da divisão entre {num1} e {num2} é: {num1 % num2}\n")
print(f"A divisão inteira entre {num1} e {num2} é: {num1 // num2}\n")
```

```python
producao_segunda = 1240
producao_terca = 1315
producao_quarta = 1288
producao_quinta = 1360
producao_sexta = 1297

producao_total = producao_segunda + producao_terca + producao_quinta + producao_quarta + producao_sexta

media_diaria = producao_total / 5

falta_meta_semanal = 7000 - producao_total

print(f"A produção total da semana foi de: {producao_total} peças")
print(f"A média diária foi de {media_diaria:.2f} peças produzidas")
print(f"Faltaram {falta_meta_semanal} peças para bater a meta de 7000 peças nesta semana")
```
