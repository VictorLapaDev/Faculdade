# Criptografia Simétrica

```python
#!pip install cryptography
from cryptography.fernet import Fernet
```

```python
chave = Fernet.generate_key()
print(chave)
```

```python
cifra = Fernet(chave) # Ação de cifrar e decifrar

mensagem = "Olá, tudo bem?" # Mensagem que será cifrada

mensagem_cifrada = cifra.encrypt(mensagem.encode()) # Cifra a mensagem, transformando em bytes

print(mensagem_cifrada)
```

```python
mensagem_decifrada = cifra.decrypt(mensagem_cifrada) # Decifra a mensagem, transformando em bytes
print(mensagem_decifrada) # Exibe a mensagem decifrada em bytes
print(mensagem_decifrada.decode()) # Decodifica a mensagem de bytes para string
```

# Criptografia Assimétrica

```python
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import hashes    
```

```python
chave_privada_gama = rsa.generate_private_key(
    public_exponent=65537,
    key_size=2048
)  

chave_publica_gama = chave_privada_gama.public_key()
```

```python
dados = "Mensagem secreta"

dados_cifrados = chave_publica_gama.encrypt(
    dados.encode(),
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)   
```

```python
dados_recuperados = chave_privada_gama.decrypt(
    dados_cifrados,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)   
```

```python
print(dados_recuperados.decode()) # Decodifica a mensagem de bytes para string
```
