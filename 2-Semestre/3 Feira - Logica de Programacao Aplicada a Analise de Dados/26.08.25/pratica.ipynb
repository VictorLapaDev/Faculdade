# Matriz

```python
estoque = [18, 5, 0, 32, 4, 7, 0, 14, 3]

quant_produtos = len(estoque)
esgotado = 0
critico = 0

for produto in estoque:
    
    if produto == 0:
        esgotado += 1
        
    if produto <= 5 and produto > 0:
        critico += 1
    
need_repo_total = critico + esgotado
need_repo_porc = 100 * need_repo_total / quant_produtos

print(f"Há {esgotado} produtos esgotados")
print(f"Há {critico} produtos em estoque critico\n")
print(f"A partir do total de {quant_produtos} produtos, {esgotado + critico} precisam de reposição, o que corresponde a {need_repo_porc:.2f}% do estoque.")
```

```python
desempenho = [
    [120, 135, 128, 141],  # Unidade 1
    [98, 105, 110, 103],   # Unidade 2
    [150, 145, 160, 155],  # Unidade 3
    [116, 161, 108, 121]   # Unidade 4
]

# Total de atendimentos por unidade
soma_unidades = []
for i in range(len(desempenho)):
    soma = sum(desempenho[i])
    soma_unidades.append(soma)
    print(f"A unidade {i+1} possui um total de {soma} atendimentos")

print("")

# Inicialização das variáveis de controle
maior_total = soma_unidades[0]
unidade_maior_total = 1

maior_semanal = desempenho[0][0]
unidade_maior_semana = 1
semana_maior = 1

soma_abaixo = 0  # Total geral de semanas abaixo do limite (110)

# Processamento do desempenho e estatísticas
for i in range(len(desempenho)):

    # Identifica a unidade com maior total acumulado
    if soma_unidades[i] > maior_total:
        maior_total = soma_unidades[i]
        unidade_maior_total = i + 1

    abaixo_unidade = 0  # Contador de semanas abaixo de 110 para a unidade atual

    # Análise das semanas individuais da unidade
    for j in range(len(desempenho[i])):
        # Contagem de semanas com desempenho abaixo de 110
        if desempenho[i][j] < 110:
            abaixo_unidade += 1
            soma_abaixo += 1

        # Registro do maior pico semanal de atendimentos
        if desempenho[i][j] > maior_semanal:
            maior_semanal = desempenho[i][j]
            unidade_maior_semana = i + 1
            semana_maior = j + 1

    print(f"Na unidade {i+1} há {abaixo_unidade} registros semanais abaixo de 110")

print(f"\nHá {soma_abaixo} registros semanais abaixo de 110 atendimentos no total.")
print(f"A unidade {unidade_maior_total} teve o maior total de atendimentos: {maior_total}.")
print(f"O maior pico semanal foi na Semana {semana_maior} da Unidade {unidade_maior_semana}, com {maior_semanal} atendimentos.")
```

```python
produtos = [
    {
        "Produto": "Notebook",
        "Valor": 4200,
        "Estoque": 8
    },
    {
        "Produto": "Mouse",
        "Valor": 120,
        "Estoque": 25
    },
    {
        "Produto": "Monitor",
        "Valor": 1350,
        "Estoque": 4
    },
    {
        "Produto": "Teclado",
        "Valor": 280,
        "Estoque": 15
    },
    {
        "Produto": "Headset",
        "Valor": 450,
        "Estoque": 3
    }
]

reposicao = 0
for produto in produtos:    
    if produto["Estoque"] < 10:
        print(f"O produto {produto["Produto"]} tem o estoque {produto["Estoque"]} que é inferior a 10")
        print(f"{produto["Produto"]} - Estoque: {produto["Estoque"]}")
        reposicao += 1
print(f"\n{reposicao} produtos necessitam de reposição")
    
```

```python
entregas = [
    {
        "Motorista": "Bruno",
        "Distancia": 32,
        "Tempo": 48,
        "Status": "Entregue"
    },
    {
        "Motorista": "João",
        "Distancia": 45,
        "Tempo": 72,
        "Status": "Atrasada"
    },
    {
        "Motorista": "Pedro",
        "Distancia": 51,
        "Tempo": 85,
        "Status": "Atrasada"
    },
    {
        "Motorista": "José",
        "Distancia": 49,
        "Tempo": 81,
        "Status": "Atrasada"
    }
]

total_entrega = 0
atrasada = 0
maior_tempo = entregas[0]["Tempo"]
motorista = ""

for entrega in entregas:
    total_entrega += 1
    
    if entrega["Status"] == "Atrasada":
        atrasada += 1
        print(f"Motorista: {entrega["Motorista"]}, Distância: {entrega["Distancia"]}, Tempo: {entrega["Tempo"]} minutos, Status: {entrega["Status"]}")
        
    if entrega["Tempo"] > maior_tempo:
        maior_tempo = entrega["Tempo"]
        motorista = entrega["Motorista"]
            
porcentagem = atrasada * 100 / total_entrega

print(f"A porcentagem de entregas atrasadas referente ao total é de {porcentagem}%")
print(f"A entrega com maior tempo foi a de {motorista}, com um tempo de {maior_tempo} minutos")
            
```
